Linked-list implemntation of queue class

Review of queues
- FIFO: first in, first out
- represent the front of the queue as the head of the linked list bc can add/remove in O(1) time (without traversing entire linked list to find the tail (if it doesn't exist) or find element before tail (even if tail pointer does exist bc we can't go backward to previous node))
    - we use a tail instance variable but we still want the next element to be removed at the front of the list (since it's FIFO, we will add elements to the back, but for stacks we added elements to the front bc it was LIFO), so we represent the front of the queue with the head of the linked list

```python
class Empty(Exception):
    pass

class LinkedQueue: # an instance of the StackQueue is a queue representation itself
    # front is aligned with the head of the list
    '''FIFO queue implementation using a singly linked list for storage.'''

    class _Node:
        '''Lightweight, nonpublic class for storing a singly linked node. Identical to the one for LinkedStack._Node'''
        __slots__ = '_element', '_next'
        def __init__(self, element, next):
            self._element = element
            self._next = next

    # init
    # length
    # is_empty
    # first
    # dequeue
    # enqueue

    def __init__(self):
        '''Initialize a new queue. / Create an empty queue.'''
        self._head = None
        self._size = 0 # number of queue elements
        self._tail = None # keep a tail too, back of queue, where we will put next element when add to queue

    def __len__(self):
        '''Return number of elements in queue.'''
        return self._size

    def is_empty(self):
        '''Return True if the queue is empty.'''
        return self._size == 0

    def first(self):
        '''Return (but do not remove) first element in the queue.

        Raise Empty exception if no elements in the queue.
        '''
        if self.is_empty():
            raise Empty('Queue is empty')
        return self._head._element # return the actual element at the front of the queue NOT the node obj but the reference to the actual element

    def dequeue(self):
        '''Remove and return first element in the queue.

        Raise Empty exception if the queue is empty.
        '''
        if self.is_empty():
            raise Empty('queue is empty')
        element = self._head._element # get current head ELEMENT, NOT the node object, that's what we want to return
        self._head = self._head._next # reassign head
        self._size -= 1 # decrement size

        if self.is_empty(): # in case that element was also the tail and now the queue is empty
            self._tail = None

        return element # return former head

    def enqueue(self, e):
        '''Add element e to the back of the queue.'''
        element = self._Node(e, None)
        if self.is_empty(): # encapsulation to call is_empty function instead of referring to self._size directly (especially relevant if using is_empty() outside of the class)
            self._head = element # assign to both head and tail
            self._tail = element
        else:
            # just reassign tail
            self._tail._next = element # assign next of current tail to new element to keep the chain connected
            self._tail = element # reassign tail
        self._size += 1 # increment queue size

        # node will be new tail node so could put self._tail = element at the end instead of putting in both if-else branches
```

Textbook notes
- worst-case O(1) time for all operations (instead of enqueue and dequeue having amortized (O(1) like with array-based implementation bc of need to occassionally resize array)
- perform operations on both ends of queue -> maintain _head and _tail reference as instance variables for each queue
- front of queue = head of list
- back of queue = tail of list
- enqueue elements at back, dequeue from front
    - bc can't efficiencly remove elements from tail of singly linked list (since can't efficiently access node preceding tail), need to keep front of queue (next element to be removed) at the head

Comparison to LinkedStack
- dequeue is similar to pop (remove head of LL)
- queue must maintain _tail reference, unlike stack which only has _head reference
- operation at head has no effect on tail, but when dequeue invoked on queue with one element, we are removing the head and tail of list simultaneously -> need to set self._tail = None for consistency
- for enqueue, newest node always becomes tail, but if new node is only node in list, then that new node also becomes the head

Runtime analysis
- all operations run in worst-case constant time
- space usage linear to current number of elements
