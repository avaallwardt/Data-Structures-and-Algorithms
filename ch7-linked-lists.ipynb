Linked-list implemntation of queue class

Review of queues
- FIFO: first in, first out
- represent the front of the queue as the head of the linked list bc can add/remove in O(1) time (without traversing entire linked list to find the tail (if it doesn't exist) or find element before tail (even if tail pointer does exist bc we can't go backward to previous node))
    - we use a tail instance variable but we still want the next element to be removed at the front of the list (since it's FIFO, we will add elements to the back, but for stacks we added elements to the front bc it was LIFO), so we represent the front of the queue with the head of the linked list

```python
class Empty(Exception):
    pass

class LinkedQueue: # an instance of the StackQueue is a queue representation itself
    # front is aligned with the head of the list
    '''FIFO queue implementation using a singly linked list for storage.'''

    class _Node:
        '''Lightweight, nonpublic class for storing a singly linked node. Identical to the one for LinkedStack._Node'''
        __slots__ = '_element', '_next'
        def __init__(self, element, next):
            self._element = element
            self._next = next

    # init
    # length
    # is_empty
    # first
    # dequeue
    # enqueue

    def __init__(self):
        '''Initialize a new queue. / Create an empty queue.'''
        self._head = None
        self._size = 0 # number of queue elements
        self._tail = None # keep a tail too, back of queue, where we will put next element when add to queue

    def __len__(self):
        '''Return number of elements in queue.'''
        return self._size

    def is_empty(self):
        '''Return True if the queue is empty.'''
        return self._size == 0

    def first(self):
        '''Return (but do not remove) first element in the queue.

        Raise Empty exception if no elements in the queue.
        '''
        if self.is_empty():
            raise Empty('Queue is empty')
        return self._head._element # return the actual element at the front of the queue NOT the node obj but the reference to the actual element

    def dequeue(self):
        '''Remove and return first element in the queue.

        Raise Empty exception if the queue is empty.
        '''
        if self.is_empty():
            raise Empty('queue is empty')
        element = self._head._element # get current head ELEMENT, NOT the node object, that's what we want to return
        self._head = self._head._next # reassign head
        self._size -= 1 # decrement size

        if self.is_empty(): # in case that element was also the tail and now the queue is empty
            self._tail = None

        return element # return former head

    def enqueue(self, e):
        '''Add element e to the back of the queue.'''
        element = self._Node(e, None)
        if self.is_empty(): # encapsulation to call is_empty function instead of referring to self._size directly (especially relevant if using is_empty() outside of the class)
            self._head = element # assign to both head and tail
            self._tail = element
        else:
            # just reassign tail
            self._tail._next = element # assign next of current tail to new element to keep the chain connected
            self._tail = element # reassign tail
        self._size += 1 # increment queue size

        # node will be new tail node so could put self._tail = element at the end instead of putting in both if-else branches
```

Textbook notes
- worst-case O(1) time for all operations (instead of enqueue and dequeue having amortized (O(1) like with array-based implementation bc of need to occassionally resize array)
- perform operations on both ends of queue -> maintain _head and _tail reference as instance variables for each queue
- front of queue = head of list
- back of queue = tail of list
- enqueue elements at back, dequeue from front
    - bc can't efficiencly remove elements from tail of singly linked list (since can't efficiently access node preceding tail), need to keep front of queue (next element to be removed) at the head

Comparison to LinkedStack
- dequeue is similar to pop (remove head of LL)
- queue must maintain _tail reference, unlike stack which only has _head reference
- operation at head has no effect on tail, but when dequeue invoked on queue with one element, we are removing the head and tail of list simultaneously -> need to set self._tail = None for consistency
- for enqueue, newest node always becomes tail, but if new node is only node in list, then that new node also becomes the head

Runtime analysis
- all operations run in worst-case constant time
- space usage linear to current number of elements

## 7.3: Doubly Linked Lists

Singly LL
- each node maintains reference to node immediately after it
- useful to manage a sequence of elements
- limitations
    - can efficienntly insert node at either end
    - can delete node from end
    - can't efficiently delete node from tail (bc no efficient way to access node previous to tail to reassign tail member w/o iterating over whole list)
    - can't efficiently delete arbitrary node from interiro position if only given reference to node
        - can't determine node that immediately precedes node to be deleted to update its ._next reference

Doubly LL
- provides greater symmetry
- each node keeps explicit reference to node before + after it
- allows more O(1)-time update operations
    - insertions + deletions at arbitrary positions within list
    - next = reference to node that follows a node
    - prev = reference to node that precedes a node

Header and Trailer Sentinels
- sentinels / guards = dummy nodes that don't store elements of primary sequence
    - header node = at beg of list
    - trailer node = at end of list
- help us avoid special cases operating near boundaries of a doubly LL (e.g. always inserting/removing node that is btwn 2 nodes bc end nodes are still between sentinels)
- initializing empty list
    - header._next = trailer
    - trailer._prev = header
    - other fields of header + trailer are irrelevant (presumably None)
- nonempty list
    - header._next = node with first real element of seq
    - trailer._prev = node with last element of seq

Advantages of Sentinels
- simplified logic of operations
- only nodes btwn header + trailer change
- treat all insertions in unified manner (new node always placed btwn 2 existing nodes)
- treat all deletions in unified manner (every element to be deleted is guaranteed to be stored btwn 2 nodes)
- LinkedQueue
    - enqueue: add new node to end of list
        - general case: new node linked after existing tail
        - insert into empty list: no existing tail, must reassign self._head to reference new node
        - using sentinel node would eliminate special case (bc always existing node before new node)

Inserting and Deleting with a Doubly LL
- every insertion takes place btwn existing nodes
- insert at front: add new node between header + node currently after header
- delete node: two neighbors to be deleted are linked, bypassing original node
    - same implementation when deleting first + last element of sequence bc element always stored at node btwn 2 others

### 7.3.1: Basic Implementation of Doubly LL
- _DoublyLinkedBase = preliminary implementation of doubly LL
    - leading underscore bc don't intend for coherent public interface for general use
    - LLs support general insertion + deletion in O(1) worst-case time only if location of operation can be sufficiently indexed
    - array-based seqs: integer index convenient for describing position in deq
    - LLs: index not convenient bc no efficient way to find jth element, would have to traverse portion of the list
- alternative to index for LL: most direct way to describe location of an operation is to identify relevant node in list
    - but want to encapsulate inner workings of data structure to avoid users directly accessing nodes of list
- 2 public classes that inherit from _DoublyLinkedBase
    - LinkedDeque (7.3.2): support for double -ended queue ADT that only supports operations at end of queue (no need to identify interior positiosn within list)
    - PositionalList (7.4): provides public interface that allows arbitrary inserations + deletions from list

Make _DoublyLinkedBase
- nonpublic _Node class (similar to singly LL)
    - _next, _element, and _prev attributes

```python
class _DoublyLinkedBase:
    """A base class providing a doubly linked list representation."""

    class _Node:
        """Lightweight, nonpublic class for storing a doubly linked node."""
        __slots__ = '_next', '_element', '_prev' # streamline memory
        def __init__(self, element, prev, next):
            self._element = element # element
            self._next # previous node reference
            self._prev = prev # next node reference

    # constructur instantiates 2 sentinel nodes + links them to each other
    # _size member
    # support for __len__ and is_empty (not a python special method) -> directly inherited by subclasses
    def __init__(self):
        """Create an empty linked list."""
        header = self._Node(None, None, None)
        trailer = self._Node(None, None, None)
        header._next = trailer
        trailer._prev = header
        self._size = 0 # header + trailer (sentinel nodes) don't contribute to size of LL

    def __len__(self):
        """Return the number of elements in the linked list."""
        return self._size

    def is_empty(self):
        """Return True if the linked list is empty."""
        return self._size == 0

    # 2 nonpublic utilities: _insert_between, _delete_node
    def _insert_between(self, e, predecessor, successor):
        """Add element e between 2 existing nodes + return the new node."""
        addition = self._Node(e, predecessor, successor) # create new node
            # use self._Node bc ._Node is an attribute of the _DoublyLinkedBase class
        predecessor._next = addition
        successor._prev = addition
        size += 1 # increment size
        return addition

    def _delete_node(self, node):
        """Delete nonsentinel noed from the list and return its element."""
        element = node._element # store element before deleting
        node._prev._next = node._next # link previous node to following node
        node._node._prev = node._prev # link following node to previous node
        # deprecate node (already stored element first)
        node._element = node._next = node._prev = None # chained assignment is valid, python evaluates right side once and assigns it to all vars
        size -= 1# decrement size
        return element # return deleted element
```

### 7.3.2: Implementing a Deque with a Doubly LL
- deque ADT
    - array-based implementation: all ops are amortized O(1) time, due to occasional need to resize array
    - doubly LL implementation: all ops in _worst-case_ O(1) time
- LinkedDeque class = deque ADT
    - inherits from _DoublyLinkedBase class
    - no explicit __ init __ method bc inherited version sufficies to initialize new instance
    - rely on __ len __ and is_empty inherited method
    - use sentinels
        - first element of deque = node after first element (if deque nonempty)
        - last element of deque = node before trailer
    - insert at either end of list = inherited _insert_between method
        - insert at front: place node immediately between header + node after header
        - insert at end: place node immediately before trailer node
            - and we have the _prev reference of trailer to help us with the linking! yay!
        - succeed even when deque empty (node placed btwn 2 sentinels)
    - delete element from nonempty deque = inherited _delete_node method
        - node assured to have neighbots on both sides

```python
class LinkedDeque(_DoublyLinkedBase): # inheritance!

    def first(self):
        """Return (but do not remove) the element at the front of the deque."""
        if self.is_empty(): # can call inherited methods!
            raise Empty('Deque is empty.')
        return self.header._next._element # return element not the node!

    def last(self):
        """Return (but do not remove) the element at the back of the deque."""
        if self.is_empty():
            raise Empty('Deque is empty.')
        return self.trailer._prev._element

    def insert_first(self, e):
        """Add an element to the front of the deque."""
        # new = self._Node(e, self.header._next._prev, self.header)
        # self.header._next._prev = new
        # self.header._next = new
        # use parent class' _insert_between
        self._insert_between(e, self.header, self.header._next)

    def insert_last(self, e):
        """Add an element to the back of the deque."""
        self._insert_between(e, self.trailer._prev, self.trailer)

    def delete_first(self):
        """Remove and return the element from the deque.

        Raise Empty exception if the deque is empty.
        """
        if self.is_empty():
            raise Empty('Deque is empty.')
        self._delete_node(self.header._next)

    def delete_last(self):
        """Remove and return the element from the back of the deque.

        Raise Empty exception if the deque is empty.
        """
        if self.is_empty():
            raise Empty('Deque is empty.')
        self._delete_node(self.trailer._prev)
```

# 7.4: The Positional List ADT

Abstract Data Types (ADTs) so far
- stacks
- queues
- deques
- have only allowed update operations that occur at one end of sequence of another

Want a more general abstraction
- queue ADT is too limiting for a customer queue
- what if a waiting customer hangs up before reaching front of queue? (removal from middle of queue)
- what if someone in line lets a friend cut? (addition to middle of queue)
- **goal**: ADT that provides user a way to refer to elements anywhere in a sequence + to perform arbitrary insertions + deletions

Array-based sequences (e.g. python list)
- integer indices decribe location of element / location for insertion or deletion

Why indices don't work for LLs
- can't efficiently access an entry knowing only index
    - finding element at index requires traverse list incrementally from beg to end (count number of elements to find desired index)
- index of entry changes over time due to insertions/deletions earlier in sequence (not convenient way to identify elements in the LL bc the indices are dynamically changing)

Example
- text document = long sequence of characters
- cursor = abstraction that describes a position within the document w/o explicit use of integer index
    - allows insert and delete operations at arbitrary positions
    - can refer to an inherent position within a document (e.g. beginning of a section) w/o relying on character index that may change as document evolves

A Node Reference as a Position?
- benefit of LL structure: can perform O(1)-time insertions + deletions at arbitrary positions of list as long as given reference to relevant node of list
- can develop an ADT where node refernece serves as mechanism for describing position
    - _DoublyLinkedBase class already has _insert_between and _delete_node methods that accept node references as parameters
- directly using methods violates object-oriented design principles of abstraction + encapsulation
- reasons for encapsulating nodes of a LL (for programmer + user)
    1) simpler for users of the DS to not be bothered with unnecessary details of our implementation (e.g. low-level manipulation of nodes or reliance on sentinel nodes)
        - _insert_bewteen method requires header sentinel as a parameter to add a node to beginning of a sequence, we don't want users of the DS dealing with sentinel nodes
    2) can provide more robust DS if we don't permit users to directly access or manipulate nodes
        - ensure users can't invalidate the consistency of a list by mismanaging linking of nodes
        - if user could directly call _insert_between or _delete_node, they could send a node that doesn't belong to the given list as a parameter
    3) encapsulate internal details of implementation -> greater flexibility to redesign the DS + improve its performance
        - well-designed abstraction -> can provide a notion of non-numeric position, even if using an array-based sequence

Solution
- don't rely on nodes
- independent **position** abstraction to denote location of an element within a LL
- complete **positional list ADT** can encapsulate a doubly LL (or even an array-based sequence -- Exerc P-7.46)

### 7.4.2 Doubly linked List Implementation

- complete implementation of a PositionalList class using a doubly LL that satisfies Proposition 7.2
- **Proposition 7.2**: Each method of the positional ADT runs in worst-case O(1) time when implemented with a doubly linked list
- rely on _DoublyLinkedBase class for low-level representation
    - new class provides a public interface for positional list ADT

PositionalList class
- nested public Position class
    - Position instance = used to represent location of elements within list
    - PositionList methods may create redundant Position instances that reference same underlying node (e.g. when first and last are the same)
    - so Position class defined __ eq __ and __ ne __ special methods so that a test to p == q evaluates to True when 2 positions refer to the same node

#### Validating Positions
- when a PositionalList method accepts a position as a parameter -> want to verify position is valid + if it is detemrine the underlying node associated with the position
    - nonpublic method _validate does this
    - position internally maintains a reference to the associated node of the LL + a reference to the list instance that contains the specific node
    - with the container reference, we can robustly detect when a caller sends a position instance that doesn't belong to the indicated list
- can also detect a position instance that belongs to the list but refers to a node no longer part of list
    - _delete_node of base class sets previous and next references of a deleted node to None -> recognize that condition to detect a deprecated node

#### Access and Update Methods
- access methods
    - trivially adapt the underlying doubly LL implementation to support public interface of positional list ADT
    - rely on _validate utility to "unwrap" any position sent
    - rely on _make_position utility to "wrap" nodes as Position instances to return to user + make sure never retur a position referencing a sentinel
    - override inherited _insert_between utility method so that ours returns a position associated with newly created node (instead of the node itself)

```python
class PositionalList(_DoublyLinkedBase):
    """A sequential container of elements allowing positional access."""
    #----------------------- nested Position class -----------------------
    class Position:
        """An abstraction representing the location of a single element."""

        def __init__(self, container, node):
            """Constructor should not be invoked by the user."""
            self._container = container # container is the LL it belongs to
            self._node = node

        def element(self):
            """Return the element stored at this position"""
            return self._node._element

        def __eq__(self, other):
            """Return True if other is a Position representing the same location."""
            return type(other) is type(self) and other._node is self._node
            # refer to same node in memory

        def __ne__(self, other):
            """Return True if the other does not represent the same location."""
            return not (self == other) # == automatically uses __eq__ when defined
                # opposite of __eq__

    #----------------------- utility method -----------------------
    def _validate(self, p):
        """Return position's node, or raise appropriate error if invalid."""
        if not isinstance(p, self.Position): # if p isn't an instance of self.Position class
            raise TypeError('p must be proper Position type')
        if p._container is not self: # if the LL this position belongs to isn't the current LL object (self)
            raise ValueError('p does not belong to this container.')
        if p._node._next is None: # convention for deprecated nodes (no longer part of LL)
            raise ValueError('p is no longer valid')
        return p._node

    #----------------------- utility method -----------------------
    def _make_position(self, node):
        """Return Position instance for given node (or None if it's a sentinel node)."""
        if node is self._header or node is self._trailer:
            return None # sentinel nodes don't have positions
        else:
            return self.Position(self, node) # passing self here passes self to the 2nd parameter which is the container! bc the first self is always automatically passed! yay!

    #----------------------- accessors -----------------------
    def first(self):
        """Return the first POSITION in the list (or None if list is empty)"""
        # list automaticlaly has sentinel nodes bc it inherited from _DoublyLinkedBase
        return self._make_position(self._header._next)
        # call _make_position instead of invoking Position constructor directly bc want to check if the node is a sentinel node and throw an error if it is!

    def last(self):
        """Return the last Position in the list (or None if list is empty)"""
        return self._make_position(self._trailer._prev)

    def before(self, p):
        """Return the Position just before Position p (or None if p is first)."""
        node = self._validate(p) # return node at the position
        return self._make_position(node._prev) # make a position for the previous node
            # if node._prev is header node, then _make_position will throw an error!

    def after(self, p):
        """Return the Position just after the Position p (or None if p is last)"""
        node = self._validate(p)
        return self._make_position(node._next)

    def __iter__(self):
        """Generate a forward iteration of the elements of the list."""
        cursor = self.first() # first position
        while cursor is not None:
            yield cursor.element()
                # cursor.element() returns the element stored at the cursor LL position
                # yield is used in generator to return the value to the caller but pause the function, saving its state.
            cursor = self.after(cursor) # position just after the position

    #----------------------- mutators -----------------------

    # override inherited version to return Position, rather than Node
    def _insert_between(self, e, predecessor, successor):
        """Add element ebwteen existing nodea dn return new Position."""
        node = super()._insert_between(e, predecessor, successor) # create the node with the super() inherited function
        return self._make_position(node) # make a position for that node and return it so that _insert_between returns a position not a node

    def add_first(self, e):
        """Insert element e at the front of the list and return new Position."""
        return self._insert_between(e, self._header, self._header._next)

    def add_last(self, e):
        """Insert element e at the back of the list and return new Position."""
        return self._insert_bewteen(e, self._trailer._prev, self._trailer)

    def add_before(self, p, e):
        """Insert element e into list before Position p and return new Position."""
        original = self._validate(p)
        return self._insert_between(e, original._prev, original)

    def add_after(self, p, e):
        """Insert element e into list after Position P and return new position."""
        original = self._validate(p)
        return self._insert_between(e, original, original._next)

    def delete(self, p):
        """Remove and return the element at Position p."""
        original = self._validate(p)
        return self._delete_node(original) # inherited method returns element

    def replace(self, p, e):
        """Replace the element at Position p with e.

        Return the element formerly at Position p.
        """
        original = self._validate(p) # returns node at position p
        old_value = original._element # temperarily store old element
        original._element = e # replace with new element
        return old_value # return old element value
```
