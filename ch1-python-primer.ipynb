### 1.1.1: The Python Interpreter
- python = interpreted language

    - interpreter received command, evaluates it, and reports result
    - programmer defines series of commands in advance + saved them in plain text file (source code or script)

    e.g. python demo.py
    - or can use an -i flag to execute a script and then enter interactive mode
        e.g. python -i demo.py
- IDE = Integrated Development Environment

### 1.1.2: Preview of a Python Program
- python syntax relies heavily on whitespace

- indentation delimits the bodies of control structures
- comment syntax: #



```python
print('Welcome to the GPA Calculator.')

# map from letter grade to point value
points = {'A+': 4.0, 'A': 4.0, 'A-': 3.67
          'B+': 3.3, 'B': 3.0, 'B-': 2.67}

num_courses = 0
total_points = 0
done = False
while not done:
    grade = input() # read line from user
    if grade == '': # empty line entered
        done = True
    elif grade not in points:
        print("Unknown grade '{0}' being ignored".format(grade))
    else:
        num_courses += 1
        total_points += points[grade]
if num_courses > 0:
    print('Your GPA is {0:.3}'.format(total_points / num_courses))
```

### 1.2.1: Identifiers, Objects, and the Assignment Statement
- assignment statement (e.g. `temperature = 98.6`)
    - establishes `temperature` as identifier (aka name)
    - associates `temperature` with the object on RHS of =
        - 98.6 = floating point value
    - the identifier temperature references an instance of the float class having value 98.6

Identifiers
- case-sensitive
- can be composed of any Unicode characters
    - cannot statr with number
    - 33 reserved words that can't be used as identifiers: False, None, True, and, as, assert, break, class, continue, def, del, elif, else, except, finally, for, from, global, if, import, in, is, lamda, nonlocal, not, or, pass, rase, return, try, while, with, yield
- similar to reference variable in Java
    - identifier is implicitly associated with memory address of object it refers to
    - may be assigned to None (like null in Java or C++)
- unlike Java and C++:
    - python is dynamically typed (no advance declaration associating identifier w/ data type)
    - identifier can be associated with any data type, can later be reassigned to another object of same or different type
    - identifier has no declared type, but object it refers to has a definite type
    - e.g. 98.6 is recognized as floating-point literal
        - so identifier `temperature` is associated with an instance of the `float` class having value 98.6

Alias: when a second identifier is assigned to an existing object
- then both identifiers (the original and the new alias) point to the same memory address (same object)
    - they are called "aliases" for the same object
- either can be used to access underlying object
- if object supports behaviors that affect its state, changes through one alias will be apparent when using other alias (bc they refer to same object)
- BUT if one of the names/aliases is reassigned to a new value using a subsequent assignemtn statement, the alias is broken (that alias no longer refers to the original object)
- e.g. 
    `temperature = 98.6`
    `original = temperature`
    `temperature = temperature + 5.0`
    - execution starts w/ evaluation of expression on RHS of =
    - `temperature + 5.0` is evaluated on existing variable `temperature`
    - result is stored as a new floating-point instance
    - then the name on the lefthand side of the assignment statement, `temperature`, is reassigned to the result
    - this does not affect the value of the existing float instance (98.6) that the identifier `original` continues to reference
        - so `temperature` now points to a new memory address (bc it's technically a new value)

### 1.2.2: Creating and Using Objects
Instantiation = process of creating a new instance of a class
- syntax generally is to invoke the "constructor" of the class
    - e.g. class named Widget -> create instance of class -> `w = Widget()` (assuming constructure requires no parameters)
- many built-in classes support "literal" form for designating new instances
    - e.g. `temperature = 98.6` creates a new instance of the `float` class and 98.6 is a literal form
- another way to indirectly create a new instance of a class: call a function that creates and returns such an instance
    - e.g. python has built-in function `sorted` that takes a sequence of comparable elements as a parameter + returns a new instance of the `list` class containing those elements in sorted order


Calling Methods
- traditional functions (e.g. `sorted(data)` where data is a parameter)
- Python classes can also define one or more "methods" (aka "member functions") invoked on a specific instance of a class using the "." operator
    - e.g. list class has a method named `sort` that can be invoked with syntax `data.sort()` which rearranges contents of `data` so they are sorted
    - expression to left of dot = identifies object upon which method is invoked
        - often an identifier (e.g. data) or the immediate result of some other operation
            - e.g. `response.lower().startswith('y')` first evaluates the method call `response.lower()` which returns a new string instance, then calls `.startswith('y')` on that intermediate string
- impt to understand behavior of a method of a class when using it
    - "accessors" = methods that return information abt state of an object but don't change the state
    - "mutators / update methods" = change the state of an object (e.g. `sort` method of `list` class)

### 1.2.3: Python's Built-In Classes
- "immutable" class = when each object of the class has a fixed value upon instantiation that cannot subsequently be changed
    - e.g. `float` class is immutable
    - once instance of class has been created, its value cannot be changed
    - though an identifier referencing the object can be assigned to a different value/object (in memory)

Python built-in classes
- immutable classes
    - bool = boolean value
    - int = integer (arbitrary magnitude)
    - float = floating-point number
    - tuple = immutable sequence of objects
    - str = character string
    - frozenset = immutable form of `set` class
- mutable classes
    - list = mutable sequence of objects
    - set = unordered set of distinct objects
    - dict = associative mapping (aka dictionary)
- some info about these classes
    - literal forms (e.g. 98.6) exist for most of them
    - all support a traditional constructor form that creates instances based upon one or more existing values (section 1.3)
    - discussed more in later chapters

The bool Class
- used to manipulate logical (Boolean) values
- 2 instances expressed as literals: `True` and `False`
- default constructor `bool()` returns False
    - easier to just use direct literal form
- can create Boolean value from nonboolean type using `bool(foo)` for value `foo`
    - interpretation depends on type of parameters
    - numbers -> evaluate to False is 0, True if nonzero
    - sequences and other container types (strings, lists) -> evaluate to False if empty, True if nonempty
    - allows you to use nonboolean value as condition in a control structure (e.g. if statement)

The int Class
- int, float = primary numeric types in Python
- int = represents integer values w/ arbitrary magnitude
    - Java, C++ = support different integral types with different precisions (e.g. int, short, long)
    - Python automatically chooses internal representation for integer based on magnitude of its value
- to express integer using binary, octal, hexadecimal = use prefix of the number 0 then character to describe the base
    - e.g. 0b1011 = binary
    - e.g. 0o52 = octal
    - e.g.  0x7f = hexadecimal
- int constructor: `int()`
    - returns 0 by default
    - can be used toconstruct integer value from existing value of another type
        - if f is floating-point value, `int(f)` produces truncated value of `f`
            - e.g. int(3.14) = 3, int(-3.9) = -3
    - can also parse a string presumed to represent an integer value
        - `int(s)` produces integer the string represents
        - if invalid string given (e.g. `int(hello)`), get ValueError raised
        - string must used base 10 by default
        - if want to conver from differnt base, can indicate as second optional parameter
            - e.g. `int('7f', 16)` evaluates to 127

The float Class
- only floating-point type in Python
- uses fixed-precision representation
    - like a double in Java or C++ rather than a float in those languages
- floating point of integral number: 2.0 or 2.
- scientific notation: 6.022e23
- constructor form float() returns 0.0
    - given parameter, attempts to return equivalent floating-point values
    - e.g. float(2) -> 2,0
    - e.g. float('3.14') -> attempts to parse string as floating point value -> rases ValueError exception

Sequence Types: list, tuple, str Classes
- sequence type = represents a collection of values in which the order is significant
- list = most general = sequence of arbitrary objects
    - like array in other languages
- tuple = immutable version of list class
    - streamlined internal representation
- str class = designed for representing immutable sequence of text characters
    - no separate class for characters (just string of length 1)

The list Class
- list instance = stores a sequence of objects
- lists
    - is a "referential structure" = stores a sequence of "references" to its elements
        - elements can be arbitrary objects (including None object)
    - "array-based" sequences, "zero-indexed" (if len=n, has elements 0 to n-1 inclusive)
    - most used container type in python, central to DS&A
- valuable behaviors
    - can dynamically expand and contract
- further explore in ch. 5
- use [] as delimiters for a `list` literal
    - [] is an empty lst
    - elements of a list need not be expressed as literals (e.g. string literal 'red'), can be identifiers (e.g. color1) if identifiers already established
- list() constructor
    - produces empty list by default
    - accepts any parameter of an **iterable** type (aka all standard container types: strings, list, tuples, sets, dictionaries)
    - e.g. `list("hello")` produces `['h', 'e', 'l', 'l', 'o']`
    - list is an iterable itself -> `list(existing_list)` constructs a new list instance referencing same ocntents at the original

The tuple Class
- provides an immutable version of a sequence -> instances have an internal representation more streamlined than list
- list = []
- tuple = ()
- to express tuple of len 1, include comma after element: (17,)
    - (17) w/o comma is viewed as simple parenthesized numeric expression

The str Class
- designed to efficiently represent an immutable sequence of characters based on Unicode intl character set
- more compact internal representations than referential lists and tuples (lists and tuples store references to objects in the iterable, while string is one reference itself)
- can use '' or ""
- **escape character** = \
    - \n for newline
    - \t for tab
    - \u for unicode characters
- also can use ''' or """ to begin/end string literal
    - advantage: newline characters are embedded naturally (instead of needing \n)
    - good for long, multiline strings of source code

The set and frozenset Classes
- set class = represents mathematical notion of set (collection of elements, no duplicates, no inherent order)
    - mutable
    - advantage of set over list: highly optimized method for checking whether a specific element is contained in the set based on DS **hash table**
    - 2 restrictions on sets due to algorithmic underpinnings:
        1) set does not maintain elements in any particular order
        2) only instances of **immutable** types can be added to a Python set (i.e. integers, floating-point numbers, character strings are eligible elements)
            - can maintain a set of tuples, but not a set of lists or set of sets bc lists + sets are mutable
- frozenset class = immutable form of set type
    - can have a set of frozensets bc can only have a set of immutable objects
- {} is delimeter for a set
    - {} represents empty dictionary not empty set
- constructor syntax set() produces empty set
    - if iterable parameter is set to constructor, then set of distinct elements of iterable is produced
    - e.g. set('hello') -> {'h', 'e', 'l', 'l', 'o'}

The dict Class
- dict class = represents **dictionary** / **mapping** from a set of distinct **keys** to associated **values**
    - i.e. map from unique student ID to student records (list of name, address, grades, etc.)
    - python implements dict w/ almost identical approach as set but with storage of associated values (makes sense bc keys are unique, no duplicates -- like items in a set)
- dict literal
    - {} produces empty dictionary
    - nonempty dictionary is comma-separates series of key:value pairs
        - e.g. {'ga': 'Irish', 'de': 'German'}
- constructor for dict class
    - accepts an existing mapping as parameter -> creates new dictionary with identical associations
    - also accepts sequence/list of key-value pairs (where pair is a tuple)
        - e.g. `dict(pairs)` where `pairs = [('ga', 'Irish'), ('de', 'German')]`

## 1.3: Expressions, Operators, and Precedence
Previously
- how names can identify existing objects (reference memory addresses)
- how literals (i.e. 98.6) and constructors (i.e. float(98.6)) can create instances of built-in python classes

- existing values can be combined into larger syntactic **expressions** using special symbols + keywords called **operators**
    - semantics of operator depends on type of operands
        - e.g. when a, b are ints, `a+b` indicates addition
        - e.g. when a, b are strings, `a+b` indicates concatenation
- **compound expressions** = rely on evaluation of 2 or more operations
    - e.g. `a+b*c`
    - order operations are evaluated in can affect overall value
    - python defines a specific order of precedence (can override by using explicit parentheses to group subexpressions)

**Logical Operators**
- **not** = unary negation
- **and** = conditional and
- **or** = conditional or 
- and and or **short-circuit**, meaning they do not evaluate the 2nd operand if the result can be determined from the value of the first (i.e. if 1st operand in and expression is false, won't evaluate 2nd. if 1s operand in or expression is true, won't evaluate 2nd)
    - useful when constructing a Boolean expression in which 2nd condition would throw error if 1st condition didn't succeed 

**Equality Operators**
2 notions of equality
1) **is**
- **is** = same identity
- **is not** = different identity
2) ==
- == = equivalent
- != = not equivalent

The difference
- `a is b` evaluates to True when identifiers a, b are aliases for the same object (carry same memory reference)
- `a == b` tests more general notion of equivalent
    - if a, b refer to sam object, then a==b should also evaluate to True
    - but a==b also evaluates to True when identifiers refer to dfferent objects that happen to have values deemed equivalent (e.g. two different 5's in memory)
    - notion of equivalence depends on data type
        - 2 strings equivalent if match character for character
        - 2 sets equivalent is have same contents (irrespective of order bc order doesn't matter w/ sets)
- usually == or != is appropriate operator
- reverse is, is not for situations when necessary to detect aliasing (2 diff identifiers pointing to / referencing same object in memory)

**Comparison Operators**
- <, <=, >, >=
- for numeric types or lexicographically (and case-sensitively) for strings
- if operands have incomparable types (e.g. 5 < 'hello'), exception raised

**Arithmetic Operators**
- +, -, *, / (true division), // (integer division), %
- +, -, * :
    - if both operands have type int, result is int
    - if >=1 operand is float, result is float
- division
    - 2 ints
        - / = **true division**
            - returns floating-point result of computation
        - // = mathematical floor of quotient (truncates decimal)
        - % = remainder of integer divison (i.e. 27 mod 4 = 3)
    - Java, C, C++ = only have / which returns truncated quotient when both operants have integral type, need at least 1 operand to be floating-point type to get true division (but still use / operator)
- negatives
    - assume q = n // m and r = n % m
        n = **dividend**
        m = **divisor**
    - python guarantees q * m + r = n
        - e.g. 6 * 4 + 3 = 27
    - when m is positive, python guarantees 0 <= r < m
    - when m is negative, python guarantees m < r <= 0 (remainder is negative)
        - e.g. -27 % 4 = -1 because -7 * 4 - 1 = -27
- same rules extend to floats (calculate remainder based on greatest int that can multiply divisor by before go past dividend)

**Bitwise Operators**
- ~ = bitwise complement (prefix unary operator)
- & = bitwise and
- | = bitwise or
- ^ = bitwise exclusive-or
- << = shift bits left, filling in with zeros
- >> = shift bits right, filling in with sign bit

**Sequence Operators**
- sequence types (str, tuple, list) support following operator syntaxes
    - s[j] = element at index j
    - s[start:stop] = slicing indices [start, stop)
    - s[start:stop:step] = slice start, start + step, start + 2*step,..., up to but not equaling stop
        - step can be negative
        - if omit start or stop, presumed to include entire rest of list on that end
    - s + t = concatenation of sequences
    - k * s = s + s + s + ... (k times)
    - val in s = check if val contained in s
    - val not in s = non-containment check
- use **zero-indexing**
- can use **negative indices** which denote distance from end of sequence
    - -1 is last element, -2 is 2nd to last element, etc.
- slices = half-open intervals with start included, stop index excluded
- lists are mutable
    - s[j] = val can be used to replace element at given index j
    - del s[j] removes element from list
    * can also use slice notation
- val in s -> checks if element equivalent to val (not memory address equivalency but value equivalency) in list
    - in strings, can check single character or substring
    - e.g. `'amp' in 'example'`
- comparison operations use **lexicographic order** and perform element by element comparison until 1st difference found
    - e.g. [5, 6, 9] < [5, 7] because entry at index 1 is less in first list
- operations supported by sequence types
    - s == t (equivalence, element by element)
    - s != t
    - s < t (lexicographically)
    - s <= t
    - s > t
    - s >= t

**Operators for Sets and Dictionaries**
Sets
- sets and frozensets support following operators:
    - key **in** s
    - key **not in** s
    - s1 == s2
    - s1 != s2
    - s1 <=, <, >=, > s2
    - s1 | s2 = union of s1 and s2
    - s1 & s2 = intersection of s1 and s2
    - s1 - s2 = set difference (set of elements in s1 but not s2)
    - s1 ^ s2 = set of elements in precisely one of s1 or s2 (not both, XOR)
- sets don't guarantee particular order so comparison operators (e.g. >) are not lexicographic -> instead based on mathematical notion of subset
    - based on subsets instead of equality
- sets also have named methods (e.g. add, remove) (see ch. 10)
Dictionaries
- do not maintain order of elements
- concept of subset doesn't make sense --> doesn't support equality operators
- d1 == d2 assesses if 2 dictionaries contain same set of key-value pairs
- d[k] for accessing a value associated with key k
- supported operators
    - d[key] (get value associated with given key)
    - d[key] = value (set or reset value associated with given key)
    - del d[key] (remove key + its value from dict)
    - key in d (containment check)
    - key not in d
    - d1 == d2
    - d1 != d2

**Extended Assignment Operators**
- extended assignment operator (e.g. +=) supported for most binary operators
- immutable type (e.g. number, string) -> syntax doesn't change value of existing object -> instead reassigns identifier to newly constructed value (new address in memory)
    - is possible for a type to redefine semantics to instead mutate the object like the `list` class does for the += operator
        - .append() on list appends single element to end of list
        - += conjoins two lists (same as .extend(list2))D
Difference in list semantics btwn `list += val` and `list = list + val`
    - a += [3] mutates the list
    - a = a + [3] creates new list, then reassigns it to identifier a




```python
# Difference in list semantics btwn `list += val` and `list = list + val`
a = [1, 2, 3]
b = a # b is new alias for a
b += [4, 5] # add [4, 5] to end of b (and a), mutate in place
b = b + [6, 7] # creates new list object of b + [6, 7] then reassigns b to that new object, a is not modified
print(a) # will be [1, 2, 3, 4, 5]
print(b) # will be [1, 2, 3, 4, 5, 6, 7]

# not the same for ints because they're immutable
# so
c = 5
d = c
d += 10 # bc ints are immutable, this creates a new int in memory (val = 15) and reassigns d to it
print(c) # will be 5
print(d) # will be 15
```

### 1.3.1 Compound Expressions and Operator Precedence
- higher precedence evaluated before lower precedence
- operators in same category: evaluated left to right (except unary operators and exponentiation are right to left)
- python allows **chained assignment** (e.g. x = y = 0) to assign multiple identifiers to same value
- also allows **chaining** of comparison operators
    - e.g. 1 <= x + y <= 10 is evaluated as (1 <= x + y) **and** (x + y <= 10) but without computing intermediate value x + y twice (yay!)

**Operator Precedence**
Highest to lowest precedence
expr = a literal, identifier, or result of previously evaluated expression
1. member access (expr.member)
2. function/method calls, subscripts/slides
3. exponentiation (**)
4. unary operators (+expr, -expr, ~expr)
5. multiplication, division
6. addition, subtraction
7. bitwise shifting (<<, >>)
8. bitwise-and (&)
9. bitwise-xor (^)
10. bitwise-or (|)
11. comparisons, containment
12. logical-not (not expr)
13. logical-and (and)
14. logical-or (or)
15. conditional (val1 if cond else val2)
16. assignments (=, +=, -=, *=, etc.)

### 1.4.1: Conditionals
