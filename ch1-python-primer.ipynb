### 1.1.1: The Python Interpreter
- python = interpreted language

- indentation delimits the bodies of control structures

```python
print('Welcome to the GPA Calculator.')

# map from letter grade to point value
points = {'A+': 4.0, 'A': 4.0, 'A-': 3.67, 'B+': 3.3, 'B': 3.0, 'B-': 2.67}

num_courses = 0
total_points = 0
done = False
while not done:
    grade = input() # read line from user
    if grade == '': # empty line entered
        done = True
    elif grade not in points:
        print("Unknown grade '{0}' being ignored".format(grade))
    else:
        num_courses += 1
        total_points += points[grade]
if num_courses > 0:
    print('Your GPA is {0:.3}'.format(total_points / num_courses))
```

### 1.2.1: Identifiers, Objects, and the Assignment Statement
- assignment statement (e.g. `temperature = 98.6`)
    - establishes `temperature` as identifier (aka name)
    - associates `temperature` with the object on RHS of =
        - 98.6 = floating point value
    - the identifier temperature references an instance of the float class having value 98.6

Identifiers
- case-sensitive
- can be composed of any Unicode characters
    - cannot statr with number
    - 33 reserved words that can't be used as identifiers: False, None, True, and, as, assert, break, class, continue, def, del, elif, else, except, finally, for, from, global, if, import, in, is, lamda, nonlocal, not, or, pass, rase, return, try, while, with, yield
- similar to reference variable in Java
    - identifier is implicitly associated with memory address of object it refers to
    - may be assigned to None (like null in Java or C++)
- unlike Java and C++:
    - python is dynamically typed (no advance declaration associating identifier w/ data type)
    - identifier can be associated with any data type, can later be reassigned to another object of same or different type
    - identifier has no declared type, but object it refers to has a definite type
    - e.g. 98.6 is recognized as floating-point literal
        - so identifier `temperature` is associated with an instance of the `float` class having value 98.6

Alias: when a second identifier is assigned to an existing object
- then both identifiers (the original and the new alias) point to the same memory address (same object)
    - they are called "aliases" for the same object
- either can be used to access underlying object
- if object supports behaviors that affect its state, changes through one alias will be apparent when using other alias (bc they refer to same object)
- BUT if one of the names/aliases is reassigned to a new value using a subsequent assignemtn statement, the alias is broken (that alias no longer refers to the original object)
- e.g. 
    `temperature = 98.6`
    `original = temperature`
    `temperature = temperature + 5.0`
    - execution starts w/ evaluation of expression on RHS of =
    - `temperature + 5.0` is evaluated on existing variable `temperature`
    - result is stored as a new floating-point instance
    - then the name on the lefthand side of the assignment statement, `temperature`, is reassigned to the result
    - this does not affect the value of the existing float instance (98.6) that the identifier `original` continues to reference
        - so `temperature` now points to a new memory address (bc it's technically a new value)

## 1.3: Expressions, Operators, and Precedence
Previously
- how names can identify existing objects (reference memory addresses)
- how literals (i.e. 98.6) and constructors (i.e. float(98.6)) can create instances of built-in python classes

- existing values can be combined into larger syntactic **expressions** using special symbols + keywords called **operators**
    - semantics of operator depends on type of operands
        - e.g. when a, b are ints, `a+b` indicates addition
        - e.g. when a, b are strings, `a+b` indicates concatenation
- **compound expressions** = rely on evaluation of 2 or more operations
    - e.g. `a+b*c`
    - order operations are evaluated in can affect overall value
    - python defines a specific order of precedence (can override by using explicit parentheses to group subexpressions)

**Logical Operators**
- **not** = unary negation
- **and** = conditional and
- **or** = conditional or 
- and and or **short-circuit**, meaning they do not evaluate the 2nd operand if the result can be determined from the value of the first (i.e. if 1st operand in and expression is false, won't evaluate 2nd. if 1s operand in or expression is true, won't evaluate 2nd)
    - useful when constructing a Boolean expression in which 2nd condition would throw error if 1st condition didn't succeed 

**Equality Operators**
2 notions of equality
1) **is**
- **is** = same identity
- **is not** = different identity
2) ==
- == = equivalent
- != = not equivalent

The difference
- `a is b` evaluates to True when identifiers a, b are aliases for the same object (carry same memory reference)
- `a == b` tests more general notion of equivalent
    - if a, b refer to sam object, then a==b should also evaluate to True
    - but a==b also evaluates to True when identifiers refer to dfferent objects that happen to have values deemed equivalent (e.g. two different 5's in memory)
    - notion of equivalence depends on data type
        - 2 strings equivalent if match character for character
        - 2 sets equivalent is have same contents (irrespective of order bc order doesn't matter w/ sets)
- usually == or != is appropriate operator
- reverse is, is not for situations when necessary to detect aliasing (2 diff identifiers pointing to / referencing same object in memory)

**Comparison Operators**
- <, <=, >, >=
- for numeric types or lexicographically (and case-sensitively) for strings
- if operands have incomparable types (e.g. 5 < 'hello'), exception raised

**Arithmetic Operators**
- +, -, *, / (true division), // (integer division), %
- +, -, * :
    - if both operands have type int, result is int
    - if >=1 operand is float, result is float
- division
    - 2 ints
        - / = **true division**
            - returns floating-point result of computation
        - // = mathematical floor of quotient (truncates decimal)
        - % = remainder of integer divison (i.e. 27 mod 4 = 3)
    - Java, C, C++ = only have / which returns truncated quotient when both operants have integral type, need at least 1 operand to be floating-point type to get true division (but still use / operator)
- negatives
    - assume q = n // m and r = n % m
        n = **dividend**
        m = **divisor**
    - python guarantees q * m + r = n
        - e.g. 6 * 4 + 3 = 27
    - when m is positive, python guarantees 0 <= r < m
    - when m is negative, python guarantees m < r <= 0 (remainder is negative)
        - e.g. -27 % 4 = -1 because -7 * 4 - 1 = -27
- same rules extend to floats (calculate remainder based on greatest int that can multiply divisor by before go past dividend)

**Bitwise Operators**
- ~ = bitwise complement (prefix unary operator)
- & = bitwise and
- | = bitwise or
- ^ = bitwise exclusive-or
- << = shift bits left, filling in with zeros
- >> = shift bits right, filling in with sign bit

**Sequence Operators**
- sequence types (str, tuple, list) support following operator syntaxes
    - s[j] = element at index j
    - s[start:stop] = slicing indices [start, stop)
    - s[start:stop:step] = slice start, start + step, start + 2*step,..., up to but not equaling stop
        - step can be negative
        - if omit start or stop, presumed to include entire rest of list on that end
    - s + t = concatenation of sequences
    - k * s = s + s + s + ... (k times)
    - val in s = check if val contained in s
    - val not in s = non-containment check
- use **zero-indexing**
- can use **negative indices** which denote distance from end of sequence
    - -1 is last element, -2 is 2nd to last element, etc.
- slices = half-open intervals with start included, stop index excluded
- lists are mutable
    - s[j] = val can be used to replace element at given index j
    - del s[j] removes element from list
    * can also use slice notation
- val in s -> checks if element equivalent to val (not memory address equivalency but value equivalency) in list
    - in strings, can check single character or substring
    - e.g. `'amp' in 'example'`
- comparison operations use **lexicographic order** and perform element by element comparison until 1st difference found
    - e.g. [5, 6, 9] < [5, 7] because entry at index 1 is less in first list
- operations supported by sequence types
    - s == t (equivalence, element by element)
    - s != t
    - s < t (lexicographically)
    - s <= t
    - s > t
    - s >= t

**Operators for Sets and Dictionaries**
Sets
- sets and frozensets support following operators:
    - key **in** s
    - key **not in** s
    - s1 == s2
    - s1 != s2
    - s1 <=, <, >=, > s2
    - s1 | s2 = union of s1 and s2
    - s1 & s2 = intersection of s1 and s2
    - s1 - s2 = set difference (set of elements in s1 but not s2)
    - s1 ^ s2 = set of elements in precisely one of s1 or s2 (not both, XOR)
- sets don't guarantee particular order so comparison operators (e.g. >) are not lexicographic -> instead based on mathematical notion of subset
    - based on subsets instead of equality
- sets also have named methods (e.g. add, remove) (see ch. 10)
Dictionaries
- do not maintain order of elements
- concept of subset doesn't make sense --> doesn't support equality operators
- d1 == d2 assesses if 2 dictionaries contain same set of key-value pairs
- d[k] for accessing a value associated with key k
- supported operators
    - d[key] (get value associated with given key)
    - d[key] = value (set or reset value associated with given key)
    - del d[key] (remove key + its value from dict)
    - key in d (containment check)
    - key not in d
    - d1 == d2
    - d1 != d2

**Extended Assignment Operators**
- extended assignment operator (e.g. +=) supported for most binary operators
- immutable type (e.g. number, string) -> syntax doesn't change value of existing object -> instead reassigns identifier to newly constructed value (new address in memory)
    - is possible for a type to redefine semantics to instead mutate the object like the `list` class does for the += operator
        - .append() on list appends single element to end of list
        - += conjoins two lists (same as .extend(list2))D
Difference in list semantics btwn `list += val` and `list = list + val`
    - a += [3] mutates the list
    - a = a + [3] creates new list, then reassigns it to identifier a




```python
# Difference in list semantics btwn `list += val` and `list = list + val`
a = [1, 2, 3]
b = a # b is new alias for a
b += [4, 5] # add [4, 5] to end of b (and a), mutate in place
b = b + [6, 7] # creates new list object of b + [6, 7] then reassigns b to that new object, a is not modified
print(a) # will be [1, 2, 3, 4, 5]
print(b) # will be [1, 2, 3, 4, 5, 6, 7]

# not the same for ints because they're immutable
# so
c = 5
d = c
d += 10 # bc ints are immutable, this creates a new int in memory (val = 15) and reassigns d to it
print(c) # will be 5
print(d) # will be 15
```

## 1.5: Functions
- **function** = traditional, stateless function that is invoked without the context of a particular class of instance of that class (e.g. `sorted(data)`)
- **method** = member function invoked upon a specific object using an object-oriented message passing syntax (e.g. `data.sort()`)
- this chapter: pure functions (methods are in ch. 2)

```python
# Syntax for defining functions
def count(data, target): # signature (established new identifier as name of function: count here, and established number of parameters it expects and names for those: data and target here) -- python is dynamically typed so signature does not designates types of parameters or type (if any) of return value
    n = 0 # n = identifier with local scope to function call
    for item in data: # item = loop variable, identifier with local scope to function call
        if item == target:
            n += 1
    return n
```

Components of a function
- **keyword def**: **signature** = establishes new identifier as name of function (e.g. count), and establishes number of parameters it expects and names for those: (e.g. data, target)
    - python is dynamically typed so signature does not designates types of parameters or type (if any) of return value
- **body** of function = indented block of code
    - each time a function is called, python creates an **activation record** that stores information relevant to current call
        - includes **namespace** to manage all identifiers that have **local scope** within the current call
        - namepsace includes function's parameters and any other identifiers defined locally within body of function
        - identifier in local scope of function caller has no relation to identifier of same name in caller's scope (though they may be set to be aliases to same object)


**Return Statement**
- used within body of function to indicate that fn should immediately case execution and return some expressed value to the caller
- if return statement executed w/o explicit argument -> None value automatically returned
    - None also returned if flow of control every reaches end of a function body without having executed return statement
- return usually final command within function
    - there can be multiple return statemnts in same function 9conditional logic controls which command is executed, if any)

### 1.5.1: Information Passing
- identifiers used to describe expected parameters are **formal parameters**
- objects sent by the caller when invoking function are **actual parameters** (me: arguments in Java)
- parameter passing in python follows semantics of standard **assignment statement**
- when function invoked, each identifier that serves as a formal parameter is assigned the respective actual parameter (provided by caller of function) in the function's local scope

```python
prizes = count(grades, 'A')
# just before function body executed, the actual parameters (grades and 'A') are implicitly assigned to formal parameters data and target (defined within local scope of count function)
data = grades # establishes the identifier data as an alias for grades
target = 'A' # establishes target as a name for the string literal 'A'

# the identifier prizes in caller's scope is assigned to the object returned within the function body
```

**Mutable Parameters**
- additional implications of parameter passing model in python when parameter is a mutuable object
- formal parameter is an alias for actual parameter
- so body of the function may interact with the object + change its state
- e.g. `data.append('F')` will add the new entry to the end of the list `data` within the function which refers to the same `data` object in memory that was passed into the function
- BUT REASSIGNING a new value to a formal parameter within a function body (e.g. `data = []`) will not alter the actual parameter. it will simply break the alias

**Default Parameter Values**
- **polymorphic function** = a function that supports more than one possible calling signature
    - can declare 1 or more default values for parameters, allowing caller to invoke function with diff number of actual parameters (if don't assign new value to parameter with default value, default value is taken)
    - e.g. `def foo(a, b=15, c=27)`
    - if caller sends 2 parameters, they're assumed to be first  and third one is default
    - can't do `bar(a, b=15, c)` with b having default value but not subsequent c -> if default parameter value present for 1 parameter, must be present for all further parameters
    - allows one of the parameters to be customized as an optional parameter (if no actual parameter passed through, the default value is taken)

Another example of polymorphic function: **built-in range function**
- technically it's a constructor for the range class, but we can treat it as a pure function
- 3 calling syntaxes supported
    1) 1-parameter form: `range(n)` generates sequence of integers from 1 to n (exclusive)
    2) 2-parameter form: `range(start, stop)` generates integers from start to end (exclusive)
    3) 3-parameter form: `range(start, stop, step)` generates integers with increments of size step rather than 1, up to but not including stop
- technically thise violates the rule that when a single parameter is sent, it serves as the stop value (instead of the start value whcih it should)
- range fixes this by:


```python
def range(start, stop=None, step=1):
    if stop is None: # if stop value not provided (only 1 value given)
        stop = start # then make the stop value that value
        start = 0 # and start as 0 as is desired
    # can reassign start and stop values to provide desired semantics
```

**Keyword Parameters**
- traditional mechanism for matching actual parameters (sent by caller) to formal parameters (declared by function signature) is based on concept of **positional arguments**
    - actual parameters sent by caller are matched in given order to formal parameters
    - e.g. `foo(a=10, b=20, c=30)` and `foo(5)` indicates a=5, while b and c are assigned default values
- python supports alternate mechanism for sending a parameter to a function -> **keyword argument** which is specified by explicitly assigning an actual parameter to a formal parameter by name
    - e.g. `foo(c=5)` assigns c=5 while a and b assigned to default value
- function author can require certain parameters ONLY be sent through keyword-argument syntax
    - several examples in python's standard libraries
    - e.g. built-in `max` function accepts keywork parameter named `key` that can be used to vary the notion of maximum being used
    - i've seen this working with pandas (i.e. keepdims=True)
`max`
- by default, it operates based on natural order of elements according to < operator for that type
- but it can be computed by comparing some other aspect of elements
    - done by providingb an auxiliary function that converts a natural element to some other value for sake of comparison
- e.g. want to to find numeric value with max magnitude (i.e. so -35 is larger than 20)
    - can use calling syntax `max(a, b, key=abs)` so that built-n abs function is sent as the valu eassociated with the keyword parameter `key`
    - when max is called, it will compare abs(a) to abs(b) rather than a to b
- max is polymorphic in number of arguments (can do max(a, b, c, d))
    - so it's not possible to designate a key function as a traditional positional element --> need to use keyword argument
- sorting functions in python support similar `key` parameter for indicating nonstandard order

### 1.7.1: Raising an Exception

    - e.g. if a function for computing square root is sent a negative value as a parameter, it can raise an exception with command `raise ValueError('x cannot be negative')`
        - raises a newly creates instance of the Valueerror class
        - error message serves as a parameter to the constructor
        - if exception is not caught within body of function, execution of the function immediately ceases + exception is propagated to the calling context (and possibly beyond)
- when checking validity of parameters sent to a function, it is customary to first verify that parameter is 1) of an appropriate type then 2) of an appropriate value

```python
# sqrt function in Python's math library performs error-checking
def sqrt(x):
    if not isinstance(x, (int, float)) # if x is not an instance of int or float classes
        raise TypeError('x must be numeric')
    elif x < 0:
        raise ValueError('x cannot be negative')
    # rest of function (after x verified to be proper type and value)
```

- use built-in function `isinstance(obj, class)` to check type of an object at run-time
    - returns True is obj is instance of class (or any subclass of that type)
    - can enter tuple of allowable types for actual class parameter
- amount of error-checking needed in a function is debatable
    - checking type and value of each parameter demands more execution time + can seem counter to nature of python if taken to extreme

```python
# implementation of sum function with rigorous error checking
def sum(values):
    if not isinstance(values, collections.Iterable): # collections.Iterable is an abstract base class that includes all of python's iterable types (e.g. list, tuple, set), collections is a module
        raise TypeError('parameter must be an iterable type')
    total = 0
    for v in values:
        if not isinstance(v, (int, float)):
            raise TypeError('elements must be numeric')
        total = total + v
    return total
```

### 1.7.2: Catching an Exception
- philosophy 1 for managing exceptional cases: **look before you leap**
    - avooid the possibility of an exception being raised with proactive conditional test


```python
# example of look before you leap: avoid possibility of exception being raised with use of a proactive conditional test
if y!= 0:
    ratio = x / y
else:
    # do smth else
```

- philosophy 2: **it is easier to ask for forgiveness than it is to get permission**
    - quote attributed to Grace Hopper
    - don't need to spend extra execution time safeguarding against every possible exceptional case as long as there is a mechanism for copying with the problem after it arises
    - this philosophy is implemented in python with the `try-except` control structure

```python
# example of try-except control structure
try:
    ratio = x / y
except ZeroDivisionError:
    # do smth else
```

**try-except control structure**
- try block is primary code to be executed, then have exception clause
- after try block, hae one or more except cases
    - each has an identified error type and indented block fo code to be executed if the designated error is raised within the try-block
- pros: non-exceptional case runs efficiently w/o extraneous checks for exceptional condition
- cons: handling exceptional case requires slightly more time w/ try-except structure than with standard conditional statement
- SO: try-except clause is best used when there is reason to believe the exceptional case is *relatively unlikely* OR when it is prohibitively expensive to proactively evaluate the condition before to avoid the exception

- exception handling is esp useful when working with user input or reading/writing to files bc interactions are less predictable
    - e.g. `fp = open('sample.txt')` = opening file with read access
        - may raise IOError for variety of reasons (e.g. nonexistent file, lack of sufficient privilege for opening file)
        - easier to atempt the command and catch resulting error than to accurately predict exactly when command will succeed
- exceptions are objects that can be examined when caught -> can establish and link an identifier to them

```python
# example of assigning identifier to exception
try:
    fp = open('sample.txt')
except IOError as e:
    print('Unable to open the file: ', e)

# e denotes the instance of the exception that was thrown
# causes detailed error message to be displayed (e.g. "file not found")
```

- try statement can handle more than one type of exception

```python
age = int(input('Enter your age in years: '))
# can fail for variety of reasons
    # call to input can raise EOFError if console input calls
    # int constructor can raise ValueError if user has not entered characters representing valid integer

# can use single except-statement to handle 2 or more types of errors in same way
age = -1 # invalid choice
while age <= 0:
    try:
        age = int(input('Enter your age in years: '))
        if age <= 0:
            print('Your age must be positive')
    except (ValueError, EOFError): # use type to designate types of errors we wish to catch with except-clause
        print('Invalid response')
        # or can write pass if don't want to print invalid response message
    # if catch error, will print response, then continue with another pass of enclosing while loop
```

- can use tuple ( , ) to designate types of errors you wish to catch with except-clause
- when error raised within try-block, remainder of that body is immediately skipped
- e.g. (above) if exception arises within call to input or call to the int constructor, then assignment to age never occurs -> immediately goes to exception clause error-handling
    - so the value of age is unchanged if erorr throws, and the loop continues

```python
# provide different responses to different types of errors: use 2 or more except-clauses as part of try-structure

age = -1
while age <= 0:
    try:
        age = int(input('Enter your age in years: '))
        if age <= 0:
            print('Your age must be positive') # this only happens if an error isn't thrown by int() or input() earlier
    except ValueError:
        print('That is an invalid age specification')
    except EOFError: # more insurmountable than simply an errant value being entered -> might want to prvide more specific error message or allow exception to interrupt the loop and propagate to higher context
        print('There was an unexpected error reading input')
        raise # re-raise the exception = allows us to provide our own response to the exception + then interrupt the while loop and propagate the exception upward
```

2 additional features of try-except structures in python
1) can have final except-clause w/o any specific identified error types using syntax `except:` to catch other possible exceptions
    - use sparingly bc difficult to suggest how to handle an error of unknown type
2) try-statement can have a `finally` clause
    - with a body of code that will always be executed in standard or exceptional cases, even when uncaught or re-raised exception occurs
    - this block is typically used for critical cleanup work (e.g. closing an open file)
