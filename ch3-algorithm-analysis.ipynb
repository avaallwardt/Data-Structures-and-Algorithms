## 3.3: Asymptotic Analysis

```python
# function to return max value in python list
def find_max(data):
  biggest = data[0]
  for val in data:
    if val > biggest:
      biggest = val
  return biggest

  '''
  Runtime
  - grows proportional to n
  - loop executes once for each data element
    - fixed number of primitives for each pass
  - runtime: O(n) where n is the number of elements in data
  '''
```
