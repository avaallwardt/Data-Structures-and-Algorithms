Challenges of Experimental analysis
1) comparability: experimental runtimes of 2 algs are difficult to compare unless experiments performed in same hardware + software environments
2) completeness: can only experiment on limited set of test inputs, omits runtimes of inputs not included in experiment

### 3.1.1: Moving Beyond Experimental Analysis

Goal of approach for analyzing efficiency of algorithms
1. evaluate efficiency independent of hardware + software environment
2. no need for implementation
3. consider all possible inputs

**Counting Primitive Operations**
- **primitive operations** = low-level instruction with constant execution time
    - assigning identifier to object
    - determining object associated with an identifier
    - arithmetic operation (e.g. addition)
    - compare 2 numbers
    - access single elementof python list by index
    - call function (excluding operations executed within function)
    - return from a function
- count number of primtitive operations executed (t) -> this number is a measure of running time
    - each primitive operation corresponds to constant number of instructions + only fixed number primitive operations -> primitive operation count correlates to actual runtime
    - assumption: runtimes of different primitive operations are fairly similar
    - t is proportional to actual runtime of alg


**Measuring Operations as a Function of Input Size**
- want to capture order of growth of algorithm's runtime
- for an algoirthm, function f(n) characterizes number of primitive operations performed as a function of input size n

**Focusing on Worst-Case Input**
- **average-case** analysis: express runtime as function of input size obtained by taking average over all possible inputs of same size
    - requires probability distn on set of inputs + sophisticated probability theory (bc some inputs are more probable than others)
- **worst case** analysis: only identify worst-case input
    - easier
    - leads to better algorithms
    - standard of success for alg to perform well in worst case -> requires it performs well on every input
