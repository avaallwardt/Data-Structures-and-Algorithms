### 8.1.1: Tree Definitions and Properties

**Tree**
- nonlinear data structure
- breakthrough in data organization
- allow us to implement many algorithms much faster than when using linear data structures (e.g. array-based lists, linked lists)
- provide natural orgnaization for data
    - ubiquitous structures in file systes, GUIs, databases, web sites, etc.
- richer organizational structure than before + after relationships of objects in a sequence
- **hierarchical** = some objects are above/below others
- terms: parent, child, ancestor, descendant

**Tree** = an ADT that stores elements hierarchically
- each element in a tree has a **parent** element (except the top element) and >=0 **children** elements
- **root** = top element of the tree / highest element

#### Formal Definition of a Tree
A **tree T** is a set of **nodes** storing elements such that the nodes have a **parent-child** relationship that satisfies the following properties:
1) If T is nonempty, it has a special node, called the **root** of T, that has no parent.
2) Each node v of T other than the root has a _unique_ **parent** node w; every node with parent w is a **child** of w

- tree can be empty (have no nodes)
- can define a tree recursively
    - tree T is either empty or consists of a node r (root of T) and a (possibly empty) set of subtrees whose roots are the children of the main root r
    - yay!

#### Other Node Relationships
- **siblings** = 2 nodes that are children of the same parent
- **external node** = node that has no children
    - aka **leaves**
- **internal node** = node that has 1 or more children

Ex: file system
- internal nodes = directories
- leaves = regular files
- root of tree = "root directory" in UNIX and Linux, represented by the symbol "/"


- node u is an **ancestor** of node v if u=v or u is an ancestor of the parent of v (so if u is anywhere up the direct chain from v to the root)
- node v is a **descendant** of node u if u is an ancestor of v
- **subtree** of T **rooted** at node v = tree consisting of all the descendants of v in T (including v itself)

#### Edges and Paths in Trees
- **edge** of tree T = pair of nodes (u, v) such that u is the parent of v, or vice versa
- **path** of tree T = sequence of nodes such that any two consecutive nodes in the sequence form an edge
    - include only 1 child at each step

Ex: inheritance relation between classes in Python program is a tree when single inheritance is used (single inheritance is when a class inherits from only 1 parent class)
- Exception classes
    - BaseException class is root of hierarchy
    - all user-defined exception classes declared as descendants of Exception class
        - descendants are everything below the node you start at, not just direct children
- general classes
    - all classes are organized into single hierarchy bc there is a built-in class `object` that is the ultimate base class
    - direct or indrect base class of all other types in python


#### Ordered Trees
- **ordered** tree = meaningful linear order among the children of each node
    - children of a node are the 1st, 2nd, 3rd, etc.
    - arrange siblings left to right according to order
    - just the children of each node need an order
    - nodes at the same level but with different parents do not need to have a collective order
- unordered tree = no particular order among subclasses of parent class

### 8.1.2: The Tree Abstract Data Type

Define a tree ADT
- use **position** as an abstraction for the node of a tree (just like with positional lists)
- element is stored at each position
- positions satisfy parent-child relationships that define the tree structure

Position object methods
- p.element() : Return the element stored at position p

Tree ADT **accessor methods** (allow user to navigate tree via positions)
- __T.root()__ : Return the _position_ (not element!) of the root of a tree T, or None if T is empty
- __T.is_root(p)__ : Return True if position p is the root of tree T
- __T.parent(p)__ : Return the position of the parent of position p, or None if p is the root of T
- __T.num_children(p)__ : Return the number of children of position p
- __T.children(p)__ : Generate an iteration of the children of position p
- __T.is_leaf(p)__ : Return True if position p does not have any children.
- __len(T)__ : Return the number of positions (+ hence elements) that are contained in tree T
- __T.is_empty()__ : Return True if tree T does not contain any positions
- __T.positions()__ : Generate an iteration of all positions of tree T
- __iter(T)__ : Generate an iteration of all elements stored within tree T
    - define special method __ iter __
        - returns an iterator object for something iterable
    - iter(obj)gives an iterator for obj
        - produces values one at a time using next()
    - generator = special kind of iterator that returns an iterator automatically using yield
        - don't define a special method, just use yield
        - allows us iterate manually with our own method
- *always generate ValueError if position is invalid for T
- notes
    - T ordered -> T.childen(p) reports children of p in natural order
    - p is leaf -> T.children(p) generates an empty iteration
    - T empty -> T.positions() and iter(T) generate empty iterations
    - sec 8.4: how to iterate through all positions of a tree
    - no methods for creating/modifying tres
        - do later when do specific implemenation of tree interface

#### A Tree Abstract Base Class in Python
- abstraction
- public interface for python ADT is often managed via __duck typing__
    - queue:
        - defined queue ADT
        - several classes implement queue interface (LinkedQueue, CircularQueue)
        - never gave a formal definition of queue ADT in python (like a general queue that all the later classes inherit)
        - all concrete implementatins were self-contained, they just happened to adhere to the same public interface
    - duck typing: if it has the methods of a queue, it can act like a queue (even if there is no formal designation that it is a queue)
- more formal mechanism to designate relationships btwn different implementations of the same abstraction (e.g. relationship btwn LinkedQueue and CircularQueue, where they are both the queue abstraction)
    - define one class that serves as an **abstract base class**
    - define **concrete classes** that inherit the abstract base class

Using an Abstract Base Class for the Tree Implementation
- Tree class = abstract base class corresponding to tree ADT
    - can provide useful code even at this high level of abstraction
    - allows more code reuse in concrete implemenations later
- Tree class has
    - nested Position class (also abstarct)
    - declarations for many accessor methods
- Tree class doesn't have
    - internal representation for storing a tree
    - 5 methods remain abstract (root, parent, num_children, children, __ len __)
        - abstract method = specifics not yet defined
        - raise a NotImplementedError
        - subclasses will override abstract methods later
            - provide concrete implementation for each behavior depending on the needs / internal representation of that specific class
- **concrete methods** defined in Tree abstract base class
    - rely on calls to abstract methods of class
    - 10 accessor methods
        - 5 left as abstract (basically means undefined, will be defined in concrete class)
        - 5 implemented based on the abstract ones
    - concrete methods will be inherited by all subclasses
        - and then when they call the abstract method, it will automatically search for the version defined in that concrete class (searches the most local scope/implementation first)
        - promotes greater code reuse (bc no need for those subclasses to reimplement the behaviors of the 5 concrete methods in the Tree class that call the abstract methods, the subclasses just need to define the abstract methods themselves)
- Tree class is abstract -> no reason to create a direct instance of it
    - exists as a base class for inheritance
    - users will only create instance of concrete subclasses

More formal approach for defining abstract base classes: abc module


```python
class Tree:
    """Abstract base class representing a tree structure."""

    # ----------------- nested Position class -----------------
    class Position:
        """An abstraction representing the location of a single element."""

        def element(self):
            """Return the element stored at this Position."""
            raise NotImplementedError('must be implemented by subclass')

        def __eq__(self, other):
            """Return True if other Position represents the same location."""
            raise NotImplementedError('must be implemented by subclass')

        def __ne__(self, other):
            """Return True if other does not represent the same location."""
            return not (self == other)  # rely on __eq__ definition

    # ----- abstract methods that concrete subclasses must support/implement -----
    def root(self):
        """Return Position representing the tree's root (or None if empty)."""
        raise NotImplementedError('must be implemented by subclass')

    def parent(self):
        """Return Position representing p's parent (or None if p is root)."""
        raise NotImplementedError('must be implemented by subclass')

    def num_children(self, p):
        """Return the number of children that Position p has."""
        raise NotImplementedError('must be implemented by subclass')

    def children(self, p):
        """Generate an iteration of Positions representing p's children."""
        raise NotImplementedError('must be implemented by subclass')

    def __len__(self):
        """Return the total number of elements in the tree."""
        raise NotImplementedError('must be implemented by subclass')

    # -------- concrete methods implemented in this class --------
    def is_root(self, p):
        """Return True if Position p represents the root of the tree."""
        return self.root() == p # call the root method for this tree and see if it equals the given position (using the special method __eq__)

    def is_leaf(self, p):
        """Return True if Position p does not have any children."""

    def is_empty(self):
        """Return True if the tree is empty."""
        return len(self) == 0
```

### 8.1.3: Computing Depth and Height

- p = position of a node
- **depth** of p = number of ancestors of p (excluding p itself)
    - depth of root is 0 (0 ancestors and position itself is excluded)
- depth of p can be defined recursively
    - If p is the root, then the depth of p is 0
    - Otherwise, the depth of p is one plus the depth of the parent of p

```python
def depth(self, p):
    """Return the number of levels separating Position p from the root."""
    if self.is_root(p): # base case is when you reach the root traversing up the tree
        return 0
    else:
        return 1 + self.depth(self.parent(p))
```

- runtime: O(dp + 1)
    - dp = depth of p in the tree T
    - constant time recursive step for each ancestor of p
    - T.depth(p) runs in O(n) worse-case time where n is the total number of positions of T
        - position of T may have depth n-1 is all nodes form a single branch
        - and in that case, you are traversing/counting all nodes
    - more informative to characterize runtime in terms of parameter dp (depth of tree) instead of n because dp is usually much smaller (usually entire tree doesn't haeve all nodes in one branch)

#### Height
- **height** of a position p in tree: (recursive definition)
    - If p is a leaf, then height of p is 0
    - Otherwise, the height of p is one more than the maximum of the heights of p's children
- height of nonempty tree T = height of root of T
- **Proposition 8.4**: The height of a nonempty tree T is equal to the maximum of the depth of its leaf positions
    - justify in Exerc R-8.3
    

```python
def _height1(self):
    """Return the height of the tree."""
    return max(self.depth(p) for p in self.positions() if self.is_leaf(p))
```

O(n^2) worst-case time
- positions() will run in O(n) time, n = number positions of T
- call depth(p) on each leaf of T
