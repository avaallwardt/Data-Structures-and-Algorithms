### 2.1.2: Object-Oriented Design Principles
Chief principles of Object-Oriented Approach
1. Modularity
2. Abstraction
3. Encapsulation

**Modularity**
- modularity = an organizing principle in which different components of a software system are divided into separate functional units
    - necessary when we have several different componennts that must interact propertly to keep system working propertly
    - keeping interactions straight requires componnents to be well-organize
- separate modules that interact in well-defined ways
    - modularity brings clarify of thought + provides natural way of organizing functions into distinct manageable units
- modularity in a software system can also provide a powerful organizing frameowkr that brings clarity to an implementation
    - in python: **module** = a collection of loosely related functions and classes that are defined together in a single file of sourc.e code
    - python standard libraries: **math module** (provides definitions for key mathematical constants, functions), **os module** (provides support for interacting with operating system), etc.
- use of modularity helps support the goals of 2.1.1    
    - increases robustness bc easier to test + debug separate components before they're integrated into larger software system
    - bugs in a complete system can be traced to particular component + fixed in relative isolation
    - structure imposed by modularity helps enable software reusaility
        - if software modules are written in general way, they can be reused when a related need arises in other contexts
        - especially relevant in study of data structures which can typically be designed with sufficient abstraction + generality to be reused in many applications

**Abstraction**
- **abstraction** = distilling a complicated system down to its most fundamental parts
    - describe parts of a system by naming them + explaining functionality
- applying abstraction paradigm to design of data structures gives rise to **abstract data types** (ADTs)
    - **ADT** = mathematical model of a data structure that specifies 
        1) type of data stored
        2) operations supported on them
        3) types of parameters of the operations
    - ADT specifies **what** each operation does, but not **how** it does it
- **public interface of an ADT** = collective set of behaviors supported by an ADT
- python treats abstractions implicitly using **duck typing**
    - python is interpreted + dynamically typed
        - interpreted language = translated into machine code at runtime by interpreter
        - compiled language = translated into machine code before executed
    - so python has no "compile time" checking of data types + no formal requirement for declarations of abstract base classes
        - instead programmer assumes an object supports a set of known behaviors when coding + interpreter raises a run-time error if those assumptions fail
    - the object actually isn't labelled/specified/checked to be of correct type until runtime -> **duck typing** bc of quote "if it walks like a duck, swims like a duck, and quacks like a duck, I'll call it a duck"
- **abstract base class (ABC)** = how python supports abstract data types
    - cannot be instantiated (cannot create instance of the class)
    - defines 1 or more common methods that all implementations of the abstraction must have
    - realize an ABC through one or more **concrete classes** that inherit from the abstract base class then provide implementations for those methods declared by the ABC
    - python: `abc` module provides formal support for ABCs (but omit declaration of `abc`)
    - will use several existing ABCs coming from `collections` module
        - includes definitions for several common data structure ADTs + concrete implementations of some of those abstractions

**Encapsulation**
- **encapsulation** = diff components of a software system should not reveal internal details of their implementations
- pros: 
    - one programmer can implement details of a component w/o concern other programmers will be writing code that intricately depends on those internal decisions (components are kept separate)
    - robustness
    - adaptability
    - allows impelemntation details of parts of a program to change w/o affecting other parts
    - easier to fix bugs, add a new functionality w/ only local changes to a component
- cons:
    - programmer of a component must maintain public interface for the component so that other programmers can write code that depends on the interface
- applied to data structures
    - some aspects are assumed to be public, others are internal details
    - convention: nonpublic names of members of a class (both data members + member functions) start with single underscore (e.g. `_secret`)
        - shouldn't rely on these outside of the class
        - these members are ommitted from documentation (bc they're only intended for internal use)
    - idea: separating functions intended for internal use in class vs. external use

### 2.2.1: Design
- design step = most impt phase in process of SWD
    - decide how to divide program's working into classes
    - decide how classes will interact, what data they will store, what actions they will perform
- rules of thumb for how to design classes
    - **Responsibilities**:
        - divide work into different actors (each actor has a responsibility)
        - describe responsibilities w/ action verbs
        - each actor is a class
    - **Independence**: 
        - define work for each class to be as independent from other classes as possible
        - subdivide responsibilities between classes so each class has autonomy over some aspect of program
        - give data (i.e. instance variables) to the class that performs actions requiring access to that data
    - **Behaviors**:
        - define behaviors for each class carefully + precisely
            - consequences of each action performed by a class should be well understood by other classes that interact with it
            - these behaviors define the methods that the class performs
        - **interface** to the class = set of behaviors for a class
            - the behaviors are how other pieces of code interact with objects from the class
- key to designing object-oriented program = defining classes, instance variables, and methods
- **CRC cards** = common tool for developing an initial high-level design for a project
    - **CRC = Class-Responsibility-Collaborator**
    - each card represents a component of program (which becomes a class)
        - top: name of component 
        - left: responsibilities for component
        - right: collaborators for component (other components that this component will have to interact with to perform its responsibilities)
    - use index cards bc each component should have small set of responsibilities and collaborators -> keeps classes manageable
- design process iterates through action/actor cycle
    - identify an action (responsibility)
    - determine an actor (component/class) best suited to perform action

Standard approach to explain and document design: use of UML diagrams
- UML (Unified Modeling Language) diagram = expresses organization of program
    - standard visual notation for expressing object-oriented software designed
    - build with computer-aided tools
- **class diagram** = one type of UML figure, 3 parts
    1) name of class (class:)
    2) recommended instance variables (fields:)
    3) recommended methods of class (behaviors:)

### 2.2.4: Testing and Debugging
- **Testing** = process of experimentally checking the correctness of program
- **Debugging** = process of tracking execution of a program + discovering errors
- most time-consuming part of program development

**Testing**
- need careful testing plan (pre-planning + design is so impt in programming + writing quality, performant code)
- verify correctness of program on representative subset of inputs
    - **method coverage**: ensure every method of class is tested at least once
    - **statement coverage**: ensure each code statement in program is executed at least once
- **special cases** = often cause programs to fail

EX: special cases for a method that sorts a sequence of integers
- sequence has 0 length (no elements)
- sequence has 1 element
- all elements sequence are same
- sequence already sorted
- sequence reverse sorted
    - to reverse string: `reverse = original[::-1]`

- consider special conditions for structures used by program
    - e.g. if use list to store data, ensure boundary cases (e.g. inserting/removing at beginning/end of list) are properly handles
- must use handcrafted test suite!
- can also run program on large collection of randomly generated inputs
    - use `random` module in python to generate random numbers, randomize order of collections, etc.

Testing Strategies
- dependencies among classes/functions of program induce a hierarchy
    - component A is above component B if A depeeds on B (e.g. function A calls function B, function A relies on parameter that is an instance of class B)
- 2 testing strategies
    1) top-down: proceeds from top to bottom of program hierarchy
        - typically used in conjunction with **stubbing** (boot-strapping technique that replaces a lower-level component with a **stub** (replacement for the component that simulates funcionality of the original))
        - allows us to isolate what we're testing (e.g. if function A calls function B to give a string, replace the call with a fixed string so that we can test A and not the return value of B)
    2) bottom-up: proceeds from lower-level components to higher-level components
        - bottom-level functions (don't invoke other functions) are tested first, then functions that only call bottom-level functions, ...
        - class that depends on no other classes tested before a dependent class
        - work our way from independent variables/functions/classes to dependent ones
        - called **unit testing**
            - functionality of a specific component is tested in isolation of the larger software project
        - pros: better isolates cause of errors to component of being tested, assuming lower-level components it relies on have been thoroughly tested

### 2.3.1: CreditCard Class
- goals of class
    - instance = credit card (have identifying information abt customer, bank, acct number, credit limit, current balance)
    - behaviors = restricts charges that would cause card's balance to exceed spending limit, no interest or late payments
- `self` = identifies the instance upon which a method is invoked (identifies the *specific* instance)
    - e.g. `my_card.get_balance()` -> identifier `self` within the definition of the `get_balance` method refers to the specific card `my_card`
        - `self._balance` refers to instance variable `balance` stored as part of that particular credit card's (e.g. `my_card`) state
        - credit card state = all the instance variable associated with that state + their values
    - in class definition `self` is an explicit parameter, but it's not referred to when caller calls the method
        - the interprteter **automatically binds** the instance that the method is invoked on to the `self` parameter

```python
# CreditCard class
# self must be passed to every member function AND the constructor
class CreditCard:
    # constructor
    def __init__(self, customer, bank, acnt, limit):
        ''' Create a new credit card instance.

        The initial balance is zero.

        customer    the name of the customer (e.g. 'John Bowman')
        bank        the name of the bank (e.g. 'California Savings')
        acnt        the account identifier (e.g. '1234 5678 9012 3456')
        limit       credit limit (measured in dollars)
        '''
        self._customer = customer
        self._bank = bank
        self._account = acnt
        self._limit = limit
        self._balance = 0 # is an instance variable but is not an argument (automatically set to 0 when instance of this class is initialized)

    # get methods (for all 5 instance variables)
    def get_customer(self):
        '''Return name of the customer.'''
        return self._customer # equivalent to this._customer in java
        # when dealing with an instance variable when defining a class, always refer to it using self. to indicate that you want the instance variable associated with the specific instance of the class that calls the method

    def get_bank(self):
        '''Return the bank's name.'''
        return self._bank

    def get_account(self):
        '''Return the card identifying number (typically stored as a string).'''
        return self._account

    def get_limit(self):
        '''Return current credit limit.'''
        return self._limit

    def get_balance(self):
        '''Return current balance'''
        return self._balance

    # more member functions / methods
    def charge(self, price):
        """ Charge given price to the card, assuming sufficient credit limit.

        Return True if charge was processed; False if charge was denied
        """
        if price + self._balance > self._limit:
            return False
        else:
            self._balance += price
            return True

    def make_payment(self, amount):
        """Process customer payment that reduces balance."""
        self._balance -= amount
```

**The Constructor**
- create instance of CreditCard class: `cc = CreditCard('John Doe', '1st bank', '1234 5678 9012 3456', 1000)`
    - results in call to `__init__` method that serves as **constructor** of the class
    - establishes state of a newly created CreditCard object w/ appropriate instance variables
    - initial values for 4/5 instance variables are provided as explicit parameters (sent by user) when instantiating credit card instance -> assigned to instance variables in body of constructor
        - e.g. `self._customer = customer` assigns instance variable `self._customer` to the parameter `customer`
        - `customer` is **unqualified** (it isn't `self._customer`), so it refers to the parameter in the local namespace


**Encapsulation**
- single leading underscore in name of data member (e.g. _balance) implies it is intended to be **nonpublic** (users of a class shouldn't be able to directly access it) (like private in Java)
- treat all data members as nonpublic
    - enforces consistent state for all instances
    - can provide access/get methods to provide user of class read-only access to a trait/instance variable
    - if want to allow user to change the state, must provide appropriate update methods
- beneficial in context of data structures bc encapsulating internal representation allows greater flexibility to *redesign* how class works / improve efficiency of a data structure

**Additional Methods**
- CC class: `charge`, `make_payment`
    - in `self._balance -= amount`, `self._balance` is **qualified** with the `self` identifier bc it represents an instance variable of the card
    - `amount` is unqualified bc it represents a local parameter

**Error Checking**
Reasons why CreditCard method isn't robust
1) Type Errors
    - do not explicitly check types of parameters to constructor, `charge`, or `make_payment`
    - if passed string to `charge`, program would crash when trying to add string to current balance
    - if class were to be widely used in a library, would want to use more rigorous techniques to raise `TypeError` when program is misued
2) Logical errors
    - user can charge a negative price, which would lower their balance (loophole for lowering balance w/o making a payment)
    - could be considered valid usage if it models credit received when returning merchandise

**Testing the Class**
- test inside the conditional `if __name__ == '__main__':` so that the tests are embedded in the source code with the class definition
- test examples in text provide **method coverage** bc all methods are called at least once but not **statement coverage** bc there is never a case in which a charge is rejected due to credit limit

### 2.3.2: Operator Overloading and Python's Special Methods
- python's built-in classes provide natural semantics for many operators
    - e.g. `a + b` invokes addition for numeric types, concatenation for sequence types
        - if a, b are lists, `a + b` returns a third list
        - a.extend(b) modifies a in place and returns node
- define new class -> must consider whether a syntax like `a+b` should be defined for instances a, b of the class

**Operator Overloading**
- default: `+` is undefined for new class
- can define operator using **operator overloading**
    - implement a specially named method
    - `+` is overloaded by implementing `__add__` which takes right hand operand (b) as a parameter, returns result of expression
    - `a+b` is converted into method call `a.__add__(b)`
- when binary operator is applied to 2 instances of different types (e.g. `3 * 'love me'`), python defers to the class of the **left** operand
    - checks if `int` class has definition for how to multiply an int instance by a string via `__mul__` method
    - if class doesn't implement such a behavior, python checks the class definition for the right-hand operand via `__rmul__` (right multiply)
- Distinction btwn `__mul__` and `__rmul__`
    - allows a new user-defined class to support mixed operations that involve an instance of an existing class
        - i.e. different definition of multiply if the instance of user-defined class is on left or right
    - allows a class to define different semantics in cases such as matrix multiplication in which operation is noncommutative (e.g. `A*x vs x*A`)


**Non-Operator Overloads**
- python also uses specially named methods to control behavior of other functionality for user-defined classes
    - e.g. `str(foo)` is formally a call to the constructor for the string class
        - if the parameter `foo` is instance of user-defined class, string constructor calls a specially named method `foo.__str__()` that must return an appropriate string representation (to see how to represent `foo` as a string)
        - basically, you call `str(foo)` and it automatically looks for `foo.__str__()` to be defined in `foo`'s class to see how to act on `foo`
        - if method doesn't exist, defaults to `__repr__()` method in user-defined class and if that doesn't exist, it uses default object representation
- also use special methods to determine how to construct `int, float, bool` based on parameter from user-defined class
    - conversion to Boolean is important to use `if foo` even when `foo` is not normally a bolean
    - `if foo` is evaluated by special method `foo.__bool__()` hopefully defined in user-defined class

More examples of top-level functions relying on specially named methods
- `len` function = determines size of container type
    - with user-defined class `len(foo)` is evaluated through call to specially named method `__len__()` of foo's class
    - often define specially named method `__len__()` when developing data structures to return a measure of the size of the structure

**Implied Methods**
- if special method is not implemented in user-defined class, standard syntax that relies on the method will raise exception
    - e.g. `a+b` for instances of user-defined class without `__add__` or `__radd__` defined will raise an error
- some operators have default definitions provided by python even in absence of special methods tho!
    - `__bool__` method (supports `if foo`)
        - automatically evaluates every object other than `None` to `True` if it's not specially defined
        - so `if foo` is saying if `foo` exists, if it isn't None
        - for containers:
            - `__len__` method is typically defined to return size of container
            - if `__len__` exists (either bc it's a container type or it's specially defined for the user-defined class), then `bool(foo)` is interpreted by default to be `True` for instances with `nonzero` length and false for instances with 0 length (using `__len__`)
            - so `if waitlist` can test whether there are one or more entries in waitlist
- python provides iterators for collections via the special method `__iter__`
    - if a container class provides implementations for both `__len__` and `__getitem__`, a default iteration is provided automatically
    - once an iterator is dfined, default functionality of `__contains__` is also provided
- `a is b` vs `a == b`
    - `a is b` checks if they're the same object in memory
        - evaluates whether identifiers are aliases for the same object (reference same object in memory)
    - `a == b` checks if they have the same characteristics, not necessarily same object in memory (i.e. a == b is true if a = 5.0 and b = 5.0)
        - notion of **equivalence** depends upon context of class
        - defined by `__eq__` method
        - if no implementation given for `__eq__`, then `a == b` is same as `a is b` (instance is equivalent to itself and nothing else)
            - if you don't define `__eq__` to override `==`, it won't compare instance varibales for you or smth it will just check if the objects are aliases for each other
- providing `__eq__` method supports `a == b` but does not affect evaluation of `a != b`
    - the `__neq__` method must be porvided
        - usually returns `not (a == b)`
- providing `__lt__` supports `a < b` and indirectly `b > a`
    - but providing `__lt__` and `__eq__` does NOT imply semantics for `a <= b`


### 2.3.3: Example: Multidimensional Vector Class
- demonstrate use of operator overloading via special methods
- better to create `Vector` class than use a list
    - list does not provide appropriate abstraction for geometric vector (adding lists concatenates two lists rather than adding components of a vector)
- relies internally on instance of a list named _coords as storage mechanism
- keep internal list encapsulated so we can enforce the desired public interface for instances of the class

Specially defined methods to override operators
- implement many behaviors trivially by invoking similar behavior on underlying list `_coords`
- implementation of `__add__` is customized
    - creates new vector + sets coordinates of new vector equal to sum of operands' elements (assuming 2 operands are vectors with same length)
    - element-by-element sum
- python **polymorphism**
    - `+` operator  executes when right-hand operand is a list of numbers with matching length bc we never declare `other` parameter must be a Vector instance, we just require that it supports `len(other)` and access to `other[j]` which both a list and a Vector object have! cool!

```python
# Vector class (abstraction for geometric vector)
class Vector:
    """Represent a vector in a multidimensional space."""

    def __init__(self, d):
        """Create d-dimensional vector of zeros."""
        self._coords = [0] * d

    def __len__(self): # specially defined method to define / override length method
        """Return the dimension of the vector."""
        return len(self._coords)

    def __getitem__(self, j):
        """Return jth coordinate of vector."""
        return self._coords[j]

    def __set__item(self, j, val);
        """Set jth coordinate of vector to given value."""
        self._coords[j] = val

    def __add__(self, other): # overrides + so we can do v1 + v2 which calls v1.__add__(v2)
        """Return sum of two vectors"""
        if len(self) != len(other): # uses specially defined length method __len__() defined earlier (bc length is not automatically defined for an object of type Vector)
            raise ValueError('dimensions must agree') # terminates program
        result = Vector(len(self))
        for j in range(len(self)):
            result[j] = self[j] + other[j]
        return result

    def __eq__(self, other):
        """Return True if vector has same coordinates as other"""
        return self._coords == other._coords # the == operator still works on the _coords list object

    def __ne__(self, other):
        """Return True if vector differs from other."""
        return not self == other # relies on existing __eq__ definition

    def __str__(self):
        """Produce string representation of vector."""
        return '<' + str(self._coords)[1:-1] + '>' # adapt list representation
```

### 2.3.4: Iterators

- rare to directly implement an iterator class
    - use **generator** syntax to automatically produce an iterator of yielded values
- python provides an automatic iterator implementation for any class that defines both
    1) `__len__`
    2) `__getitem__`
- SequenceIterator(data) class keeps an internal reference to the data sequence + current index into the sequence
    - each time `__next__` is called, the index is incremented until reaching end of sequence

```python
# implementation of an iterator
class SequenceIterator:
    """An iterator for any of Python's sequence types."""

    def __init__(self, sequence):
        """Create an iterator for the given sequence."""
        # iterator stores the sequence and current index
        self._seq = sequence # keep reference to underlying data
        self._k = -1 # index, will increment to 0 on first call to next

    def __next__(self):
        """Return the next element, or else raise StopIteration error."""
        self._k += 1 # advance to next index
        if self._k < len(self._seq):
            return (self._seq[self._k]) # return next data element
        else:
            raise StopIteration() # no more elements

    def __iter__(self):
        """By convention, an iterator must return itself as an iterator."""
        return self
```

### 2.3.5: Example: Range Class
- class that mimics python's built-in `range` class

History of built-in `range` class
- before python3, range was implemented a function -> returned list instance with elements in specified range
    - e.g. range(2, 10, 2) returned [2, 4, 6, 8]
    - but this caused instantiation + initialization of a list with the range of numbers
    - unnecessarily expensive (time + memory) (all items of list held in memory at once)
- now it uses **lazy evaluation**
    - range no longer creates a list instance
    - range is a class
    - range represents the desired range of elements without very storing them explicitly in memory
    - call `r = range(8, 140, 5)` -> result is a lightweight object (instance of range class) with only a few behaviors
        - `len(r)` -> number elements in range
        - `__getitem__(num)` -> returns specific elemnt in range
    - range class supports `__len__` and `__getitem__` -> inherits automatic support for iteration -> can execute a for loop over a range (bc range inherits support for iteration!)
    - but it doesn't actually instantiate + store every member of the range!!!

```python
class Range:
    """A class that mimics the built-in range class."""

    def __init__(self, start, stop=None, step=1):
        """Initialize a Range instance.

        Semantics is similar to built-in range class.
        """
        if step == 0:
            raise ValueError('step cannot be 0')

        if stop is None: # special case of range(n) - start should be 0 and start set to stop
            start, stop = 0, start # utilizes automatic tuple packing, unpacking

        # calculate the effective length once (more efficient)
        self._length = max(0, (start - start + step - 1) // step)

        # need knowledge of start and step (but not stop -- bc we know when length is exceeded) to support __getitem__
        self._start = start
        self._step = step

    def __len__(self):
        """Return number of entries in the range."""
        return self._length

    def __getitem__(self, k):
        """Return entry at index k (using standard interpretation if negative)."""
        if k < 0:
            k += len(self) # convert negative index using len method previously defined :)

        if not 0 <= k < self._length:
            raise IndexError('index out of range')

        return self._start + k * self._step
```

### 2.4.2: Hierarchy of Numeric Progressions
- **numeric progression** = sequence of numbers where each number depends on 1 or more previous numbers
    - **arithmetic progression** = add fixed constant to previous number
    - **geometric progression** = multiply previous value by a fixed constant
- hierarchy of progression classes
    - general base class: Progression
        - produces progression of whole numbers 0, 1, 2, ...
        - serves as base class for other progression types to maximize common functionality + minmize burden on subclasses
    - subclasses: ArithmeticProgression, GeometricProgression, FibonacciProgression

Progression class
- implements conventions of python iterator
    1) `__next__` method
    2) `__iter__` method
- each call to next(seq) returns subsequent element of progression sequence
- can also say `for value in seq` but our default progression is defined as an infinite sequence
- nonpublic method _advance updates value of self._current
    - adds 1 to the current value
    - subclasses should override _advance to provide a different rule for computing the next entry
- utility method print_progression displays the next n values of the progression


```python
# implementation of basic Progression class
class Progression:
    """Iterator producing a generic progression. (iterator has __next__ and __iter__)

    Default iterator produces the whole numbers 0, 1, 2, ...
    """
    def __init__(self, start=0):
        """Initialize current to the first value of the progression."""
        self._current = start # instance variables always referred to with start.

    def _advance(self):
        """Update self._current to a new value.

        This should be overriden by a subclass to customize progression.

        By convention, if current is set to None, this designates the end of a finite progression.
        """
        self._current += 1 # self._current is an int (immutable) so this creates a new integer object in memory

    def __next__(self):
        """Return the next element, or else raise StopIteration error."""
        if self._current is None: # our convention to end a progression
            raise StopIteration()
        else:
            answer = self._current # answer is an alias for current
            self._advance() # assigns _current to a new integer in memory
            return answer # returns the original _current object

    def __iter__(self):
        """By convention, an iterator must return itself as an iterator"""
        return self

    def print_progression(self, n):
        """Print next n values of the progression."""
        print(' '.join(str(next(self)) for j in range(n))
```

**An Arithmetic Progression Class**
- add fixed constant to one term of progression to produce the next
    - e.g. constant = 4: 0, 4, 8, 12, ...
- relies on Progression as its base class
    - constructor
        - takes in increment value and starting value
        - calls super constructor to initialize _current data member to the desired start value
        - creates new _increment data member for the arithmetic progression
    - override _advance method to add the increment value to the start value instead of adding 1 each time

```python
class ArithmeticProgression(Progression):
    def __init__(self, increment=1, start=0):
        """Create a new arithmetic progression.

        increment   the fixed constant to add to each term (default 1)
        start       the first term of the progression (default 0)
        """
        super().__init__(start) #initialize base class
        self._increment = increment

    def _advance(self): # overrides inherited version / parent class' _advance method
        # method name starts with _ to indicate it's protected, only for internal use
        """Update current value by adding the fixed increment."""
        self._current += self._increment

    # don't have to include any more methods bc they're all inherited from the parent class
    # now when the inherited __next__ method calls _advance, it will use the _advance method defined for THIS subclass
```

**A Geometric Progression Class**
- each value = fixed constant x previous value
    - fixed constant = **base** of geometric progression
    - starting point = 1 (not 0)
- e.g. base 2: 1, 2, 4, 8, 16,...


```python
class GeometricProgression(Progression): # inherit from progression
    """Iterator producing a geometric progression."""

    def __init__(self, base=2, start=1):
        """Creates a geometric progression

        base    the fixed constant multiplied to each term (default 2)
        start   the first term of the progression (default 1)
        """
        super().__init__(start)
        self._base = base

    def _advance(self):
        """Update current value by multiplying it by the base value."""
        self._current *= self._base
```

**A Fibonacci Progression Class**
- value = sum of 2 previous values
- can be generated from any 2 starting values
    - start=0, end=1 -> 0, 1, 1, 2, 3, 5, 8, ...
    - start=4, end=6 -> 4, 6, 10, 16, 26, 42, ...
- introduce new data member _prev to store value preceding the current one
- by initializing _prev to (second - first), initial advancement will set new current value to first + (second - first) = second, as desired!


```python
class FibonacciProgression(Progression):
    """Iterator producing a generalized Fibonacci progression."""

    def __init__(self, first=0, second=1):
        """Create a new Fibonacci progression.

        first   the first term of the progression (default 0)
        second  the second term of the progression (default 1)
        """
        super().__init__(first) # starting progression at first
        self._prev = second - first # fictitious value preceding the first

    def _advance(self):
        """Update current value by taking sum of previous 2"""
        self._prev, self._current = self._current, self._prev + self._current
```

### 2.4.3: Abstract Base Classes
Defining group of classes part of inheritance hierarchy
- to avoid repetition of code, design a base class with common functionality that can be inherited by other classes that need it
- e.g. Progression class is base class for 3 subclasses (ArithmeticProgression, GeometricProgression, FibonacciProgression)
    - instance of Progression class = same as ArithmeticProgression with increment 1, so we didn't really need Progression class for some functional reason
    - real purpose of Progression class was to centralize implementations of behaviors that the other subclasses needed -> streamlines code in those subclasses
- **abstract base class** = class where its only prupose is to sevre as a base class through inheritance
    - it can't / shouldn't be directly instantiated
    - it should just be the base class for other subclasses
- **concrete class** = can be instantiated
    - Progression class is technically concrete but we designed it as an abstract base class
    - subclasses of Progression are concrete classes
- in statically typed languages (Java, C++), abstract base class serves as formal type that may guarantee 1 or more **abstract methods**
    - provides support for polymorphism (one variable may have an abstract base class as its declared type even tho it refers to an instance of a concrete subclass)
    - python doesn't have declared types, so this kind of polymorphism can happen w/o need for unifying abstract base class
        - so defining abstract base classes is not as a strong a tradition in python
        - still, `abc` module provides support for defining formal abstract base class

Data Structures
- focus on abstract base classes bc Python's `collections` module provides several sbtract base classes that help hwen defining *custom data structures* that share a common inbterface with some of python's *built-in data structures*
- they rely on **template method pattern**
    - object-oriented software design pattern
    - used when an abtract base class provides concrete behaviors that rely upon calls to other abstract behaviors
    - so as soon as a subclass provides definitions for missing abstract behaviors, the inherited concrete behaviors are well defined

Example:
- collections.Sequence = abstract base class, defined behaviors common to `list`, `str`, `tuple` classes (sequences that support element access via an integer index)
    - provides concrete implementations of methods (`count`, `index`, `__contains__`) that ca be inherited by any class that provides concrete implementations of:
        1) `__len__`
        2) `__getitem__`
- implementation relies on 2 advanced python techniques
    1) declare the ABCMeta class of the abc module as a **metaclass** of our Sequene class
        - metaclass = provides a template for the class definition itself, unlike a superclass
        - ABCMeta declares assures that the constructor for the class raises an error
    2) use of @abstractmethod decorator immediately before `__len__` and `__getitem__` methods declared 
        - declares these 2 methods to be abstract (do not provide an implementation within Sequence base class but expect any concrete subclasses to support those 2 methods)
        - python will not allow instantiation for any subclass that does not override the abstract methods with concrete implementations
- implementations of other methods: `__contains__`, `index`, `count`
    - do not rely on any assumption about self instances
        - len(self) and self[j] work because of special methods `__len__` and `__getitem__`
        - support for iteration is automatic as well

```python
# implementation of Sequence abstract base class
from abc import ABCMeta, abstractmethod

class Sequence(metaclass=ABCMeta):
    """Our own version of collections.Sequence abstract base class."""

    @abstractmethod
    def __len__(self):
        """Return the length of the sequence."""

    @abstractmethod
    def __getitem__(self, j):
        """Return the elemtn at index j of the sequence."""

    def __contains__(self, val):
        """Return True if val found in the sequence; False otherwise."""
        for j in range(len(self)):
            if self[j] == val: # found match
                return True
        return False

    def index(self, val):
        """Return leftmost index at which val is found (or raise ValueError)"""
        for j in range(len(self)):
            if self[j] == val: # leftmost match
                return j
        raise ValueError('value not in sequence')

    def count(self, val):
        """Return the number of elements equal to given value."""
        k = 0
        for j in range(len(self)):
            if self[j] == val:
                k += 1
        return k
```

### 2.5.1: Instance and Class Namespaces

**How Entries are Established in a Namespace**
- _balance --> in CC instance namespace
- make_payment --> in CC class namespace
- the `__init__` method determines what namespace a member (function or variable) lives in
    - self._balance = 0
        - self = identifier for newly constructed instance
        - use of self as a qualifier for self._balance causes _balance identifier to be added to instance namespace
- when inheritance is used, there is a single instance namespace per object
    - e.g. create instance of PredatoryCC --> _apr, _balance, _limit attributes reside in instance's namespace bc they were all assigned using qualified syntax (self.)
- class namespace includes all declarations made directly within body of class definition
    - make_payment is declared within scope of CC class --> becomes associated with name make_payment within CC class namespace
    - member functions are the most typical types of entries declared in a class namespace
        - but other types of data values, including other classes, can be declared in class namespace too!

**Class Data Members**
- class-level data member = used when there is some value (e.g. constant) that is to be shared by all instances of a class
- would be wasteful to have each instance store that value in its instance namespace
- e.g. in PredatoryCC class better to assign $5 fee to a variable than embed literal value in our code
- add OVERLIMIT_FEE as a class data member
    - added to PredatoryCC namespace bc the assignment occurs
        1) within immediate scope of class definition
        2) without any qualifying identifier (self.)

```python
class PredatoryCreditCard(CreditCard):
    OVERLIMIT_FEE = 5 # class-level member

    def charge(self, price):
        success = super().charge(price)
        if not success:
            self._balance += PredatoryCreditCard.OVERLIMIT_FEE # use class name qualifying identifier instead of self. bc self. is used for instance namespace level members
        return success
```

**Dictionaries and the `__slots__` Declaration**

DEFAULT
- python represents each namespace with an instance of the built-in dict class  
    - maps identifying names in the scope to the associated object
    - dictionary structure supports relatively efficient name lookups BUT it requires additional memory usage beyond the raw data it stores

More direct mechanism for representing instance namespaces
- class definition must provide a class-level member named `__slots__` that is assigned to a fixed sequence of strings that are the instance variable names
- when inheritance is used, ,if the base class declares `__slots__`, then a subclass must also declare `__slots__` to avoid creation of instance dictionaries
    - declaration in the subclass should only include newly introduced names
- allows streamlined representation for all instances of a class
- could use `__slots__` declaration to streamline every class, but so much rigor would be atypical for python programs
    - but there are a few classes for which we expect a large number of instances but a lightweight construct
    - e.g. linked lists + trees are often comprised of many individual nodes -> use `__slots__`
- to promote greater energy efficiency, we use an explicit `__slots__` declaration in any nested classes of which we expect many instances

```python
# __slots__ example
class CreditCard:
    __slots__ = '_customer', '_bank', '_account', '_balance', '_limit'
    # the instance variable names are technically a tuple

class PredatoryCreditCard(CreditCard):
    __slots__ = '_apr'
```

### 2.5.2: Name Resolution and Dynamic Dispatch
- name resolution process = retrieving a name from namespace in python's object-oriented framework

    - other languages use **static dispatching** (make compile-time decision as to which version of function to call based on declared type of a variable)

## 2.6: Shallow and Deep Copying
- `foo = bar` makes `foo` an **alias** for object identified as `bar`
- **copy** of object = when want to modify original object vs copy independently
- `palette = warmtones` creates alias `palette` for `warmtones` list
- `palette = list(warmtones)` creates a new instance of list class and assigns it to identifier `palette` so when `warmtones` is modified, `palette` isn't also modified
    - this is a **shallow copy** (new list is initialized so its contents are precisely the same as the original sequence)
    - python lists are **referential** so the new list represents a sequence of the SAME references as `warmtones` list
    - allows us to add/remove elements from `palette` without affecting `warmtones` but editing a color instance from the palette list WILL change contents of warmtones
    - palette[0] and warmtones[0] are aliases for the same color instance
- **deep copy** = new copy references its own copies of objects referenced by original list

**Python's copy Module**
- copy module can produce both shallow copies and deep copies of arbitrary objects
- copy() function creates a shallow copy of argument
- deepcopy() function creates a deep copy of argument

```python
import copy
palette = copy.copy(warmtones) # shallow copy of warmtones list
palette = copy.deepcopy(warmtones) # deep copy of warmtones list
```
